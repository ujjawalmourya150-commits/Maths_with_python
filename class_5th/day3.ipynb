```python
# Area of a Circle Sector

Angle = int(input("Enter the angle : "))
radius = int(input("Enter the radius : "))

Area = (Angle / 360) * 3.14 *( radius**2)
print(Area)
```

```python
# Sum of Digits

num = input("Enter a number : ")
i = 0
for j in num:
    i = i + int(j)
print(f"The sum of all digits is {i}")
```

```python
# Reverse a number 

num = int(input("Enter a number: "))

reverse = 0

while num > 0 :
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print(reverse)
    
```

```python
# Check Even/Odd Without % 

num = int(input("Enter a number : "))
if num // 2 == num / 2:
    print("Even number")
else:
    print("odd number ")
```

```python
# Calculate Power Without **

base = int(input("Enter base number : "))
exponent = int(input("Enter exponent number : "))

result = 1

for i in range(exponent):
    result = result * base
print("Answer:",result)
```

```python
# Factorial

num = int(input("Enter a number : "))

factorial = 1

for i in range(1,num+1):
    factorial = factorial * i
print(factorial)
```

```python
# Prime Number Checker

num = int(input("Enter the number : "))

if num <= 1:
    print(f"{num} is not a prime number")
else:
    for i in range(2,num):
        if num % i == 0:
            print(f"{num} is not a prime number")
            break
    else:
            print(f"{num} is a prime number")
```

```python
# Fibonacci Series 
num = int(input("Enter a numeber : "))

first = 0
second = 1

for i in range(num):
    print(first)
    next_number = first + second
    first = second
    second = next_number
```

```python
# Find GCD

n1 = int(input("Enter the 1st number : "))
n2 = int(input("Enter the 2nd number : "))

while n2 != 0:
    remainder  = n1 % n2
    n1 = n2
    n2 = remainder
print(f"GCD  is {n1}")
```

```python
# Armstrong Number

num = int(input("Enter a number: "))

original_num = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total = total + digit ** digits
    num = num // 10

if total == original_num:
    print(f"{original_num} is an Armstrong number")
else:
    print(f"{original_num} is not an Armstrong number")
```
