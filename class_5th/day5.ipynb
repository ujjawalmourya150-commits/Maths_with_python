```python
# 1 . Reverse a number

num = int(input("Enter a number : "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print(reverse)
```

```python
# 2. Count digits

num = int(input("Enter a digit : "))

count = 0

while num > 0:
    num = num//10
    count +=1
print("Number of digits:", count)
```

```python
# 3 . Palindrome number

num = int(input("Enter a number : "))

original = num
i = 1
reverse = 0

while num>0:   
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10    

if original == reverse:
    print("Its a palidrome number")
else:
    print("not palidrome number")
```

```python
# 4 . Keep asking until correct

answer = True
while answer:
    num = int(input("Enter a number : "))
    if 1 <= num <= 10:
        print(f"{num} is in range of 1 to 10")
        answer = False
```

```python
# 5  .Guess the Number
import random
random_number = random.randint(1,10)

Game = True

while Game:  
    Guess = int(input("Enter a number : "))
    if Guess == random_number:
        print("You guessed right")
        Game = False
    elif Guess > random_number:
        print("Too high")
    elif  Guess < random_number:
        print("Too low")
```

```python
# 6 . Find the second largest number

numbers = [10, 25, 8, 45, 32, 18]
largest = numbers[0]
second = numbers[0]

for i in numbers:
    if largest < i:
        second = largest
        largest = i
    elif i > second and i != largest:
        second = i
print(f"{second} is the second largest ")
```

```python
# 7 . Count vowels

word = input("Enter a word : ")
vowels = ['a', 'e', 'i', 'o', 'u']
count = 0
for i in word:
    for j in vowels:
        if i == j:
            count = count + 1
print(f"Vowels in this words are {count}")            
```

```python
# 8 . Find common elements

a = [1, 2, 3, 4, 5]
b = [3, 5, 7, 9]

print("The common numbers in two list are",end=' ')

for i in a:
    for j in b:
        if i == j:
            print(i,end=' ')
```

```python
# 9. Find duplicate numbers

numbers = [2, 5, 2, 8, 5, 9, 2]

duplicates = []

for i in numbers:
    for j in numbers:
        if i == j and i not in duplicates:
            duplicates.append(i)

print(duplicates)
```
