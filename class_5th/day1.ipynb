```python
# program for area of circle

radius = float(input("Enter the radius of the circle : "))
area = 3.14159 * (radius**2)
print(f"Area of circle is {area}")
```

```python
# for area of rectangle

height = float(input("Enter height of the rectangle: "))
width = float(input("Enter weight of the rectangle: "))
area = height * width
print(f"Area of rectangle is {area} ")

```

```python
# for area of square
side = float(input("Enter side of the square: "))
area = side**2
print(f"Area of square is {area}")
```

```python
#for area of triangle
height = float(input("Enter height of the triangle: "))
base = float(input("Enter base of the triangle: "))
area =  0.5 * height * base
print(f"Area of triangle is {area}")
```

```python
# for cube root
a = int(input("enter the number : "))
cuberoot = int(a**(1/3))
print(f" Cube root of {a} is {cuberoot} ")
```

```python
# Perimeter of a Square
side = float(input("Enter the side of the Square : "))
perimeter = 4 * side
print(f"Perimeter of Square is {perimeter}")

```

```python
# perimeter of rectangle
length = float(input("Enter the lenght of the rectangle : "))
breadth = float(input("Enter the breadtth of the rectangle : "))
perimeter = 2 * (length + breadth)
print(f"Perimeter of rectangle is {perimeter} ")
```

```python
# Perimeter of a circle
radius = float(input("Enter the radius of the circle in meter : "))
perimeter = 2 * 3.14159 * radius
print(f"Perimeter of circle is {perimeter} ")

```

```python
# Average of three numbers
n1 = int(input("Enter 1st number : "))
n2 = int(input("Enter 2nd number : "))
n3 = int(input("Enter 3rd number : "))

sum = n1 + n2 + n3
average = sum / 3
print(f"The average of {n1},{n2} and {n3} is {average}")
```
