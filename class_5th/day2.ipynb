```python
# Percentage of marks

maths = int(input("enter your marks in maths : "))
hindi = int(input("enter your marks in hindi : "))
english = int(input("enter your marks in english : "))
social = int(input("enter your marks in social : "))
science = int(input("enter your marks in science : "))

total = maths + hindi + english + social + science
print(f"The total of all subject is {total} out of 500")

percentage = (total/5)*(100/100)
print(f"{percentage}%")

if percentage >= 31 and percentage <45:
    print("passes by c grade")
elif percentage >= 45 and percentage <60:
    print("passes by b grade")
elif percentage >=60 and percentage <75:
    print("passed by b+ grade")
elif percentage >= 75 and percentage <85:
    print("passed by a grade")
elif percentage >= 85 and percentage <=100:
    print("passed by a+ grade")
else:
    print("failed")
```

```python
# for negative and positive number 

num = int(input("Enter a number : "))

if num<0:
    print(f"{num} is negative")
else:
    print("positive")
```

```python
# for odd and even

num = int(input("Enter a number : "))

if num % 2== 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

```python
# for simple profit 
Sp = int(input("Enter selling price: "))
Cp = int(input("Enter cost price: "))

simple_profit = Sp - Cp
print(f"Simple profit is {simple_profit}")
```

```python
# for simple interest

P = float(input("Enter principle rate: "))
R = float(input("Enter rate of interest per year: "))
T = float(input("Enter time in years: "))

SI = (P*R*T)/100
print(f"The simple interest is {SI}")
```

```python
# for compound interest
P = float(input("Enter principle amount : "))
R = float(input("Enter Rate of Interest per year (%) :"))
T = float(input("Enter time in years : "))

A = P*(1+R/100)**T
CI = A - P

print(f"The Compound interest is  {CI}")
```

```python
# Speed calculator

distance = float(input("Enter distance: "))
time = float(input("Enter time: "))

Speed = distance / time
print(f" Speed is {Speed}")
```

```python
# BMI

Weight = float(input("Enter your weight in kg : "))
Height = float(input("Enter your height in m : "))

BMI = Weight / (Height)**2
print(f"your bmi is {round(BMI,2)}")
```
