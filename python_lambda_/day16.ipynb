```python
# 1 .  Use map() to create a list of squares

numbers = [2, 4, 6, 8, 10]

result = list(map(lambda x: x**2,numbers))

print(result)
```

```python
# 2 . # Use map() to double every number

numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 2, numbers))

print(result) 
```

```python
# 3 . # Use map() to convert every number into a string

numbers = [10, 20, 30, 40, 50]

result = list(map(lambda x: str(x), numbers))

print(result)
```

```python
# 4 . Use map() to add 5 to every number

numbers = [10, 20, 30, 40, 50]

result = list(map(lambda x: x + 5, numbers))

print(result)
```

```python
# 5 . Convert Celsius to Fahrenheit

celsius = [0, 10, 20, 30, 40]

result = list(map(lambda c: (c * 9/5) + 32, celsius))

print(result)
```

```python
# 6 .Use map() to find cubes

numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x ** 3, numbers))

print(result)
```

```python
# 7 .Use map() to find the length of each word

words = ["Python", "AI", "Machine", "Learning"]

result = list(map(lambda x: len(x), words))

print(result)
```

```python
# 8 .Use map() to convert every name to uppercase

names = ["ujjawal", "rahul", "aman", "rohit"]

result = list(map(lambda x: x.upper(), names))

print(result)
```

```python
# 9 . Use map() to check whether numbers are even or odd

numbers = [1, 2, 3, 4, 5, 6]

result = list(map(lambda x: "Even" if x % 2 == 0 else "Odd", numbers))

print(result)
```

```python
# 10 . Use map() and Give a 10% discount on every price
 
prices = [500, 1000, 1500, 2000]

result = list(map(lambda x: x - (x * 10 / 100), prices))

print(result)
```
