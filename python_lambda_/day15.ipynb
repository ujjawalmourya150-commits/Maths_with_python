```python
# 1 . Use filter() + lambda to get numbers divisible by 3 but not by 5.

numbers = [10, 12, 15, 18, 20, 21, 25, 30, 33, 40]

result = list(filter(lambda x:x % 3 == 0 and x % 5 != 0, numbers))

print(result)
```

```python
# 2 . Use filter() + lambda to get numbers greater than 20 and less than 50.

numbers = [10, 22, 35, 50, 17, 42, 55, 29, 60]

result = list(filter(lambda x:x > 20 and x < 50, numbers))

print(result)
```

```python
# 3 . Use filter() + lambda to get even numbers greater than 10.

numbers = [4, 8, 12, 15, 18, 21, 24, 30, 33]

result = list(filter(lambda x: x % 2 == 0 and x > 10, numbers))

print(result)
```

```python
# 4 . Use filter() + lambda to get numbers whose last digit is even.

numbers = [12, 25, 37, 44, 51, 68, 73, 90, 101]

result = list(filter(lambda x: x % 10 % 2 == 0, numbers))

print(result)
```

```python
# 5 . Use filter() + lambda to get words having exactly 5 characters.

words = ["apple", "cat", "hello", "python", "chair", "book", "train"]

result = list(filter(lambda x: len(x) == 5, words))

print(result)
```

```python
# 6 . Use filter() + lambda to get names that end with "n".

names = ["Aman", "Rahul", "Karan", "Rohit", "Arjun", "Priya", "Varun"]

result = list(filter(lambda x: x.endswith("n"), names))

print(result)
```

```python
# 7 . Use filter() + lambda to get numbers divisible by either 3 or 7.

numbers = [6, 10, 14, 15, 21, 25, 28, 30, 35, 40]

result = list(filter(lambda x:x % 3 == 0 or x % 7 == 0, numbers))

print(result)
```

```python
# 8 . Use filter() + lambda to get numbers that are odd.

numbers = [10, 13, 18, 21, 25, 30, 37, 42, 49]

result = list(filter(lambda x: x % 2 != 0, numbers))

print(result)
```

```python
# 9 . Use filter() + lambda to get words that start with "A" and have more than 4 characters.

words = ["Apple", "Ant", "Animal", "Amazing", "Ball", "Arrow", "AI"]

result = list(filter(lambda x:x.startswith("A") and len(x) > 4, words))

print(result)
```

```python
# 10 . Use filter() + lambda to get numbers whose last digit is 0 or 5.

numbers = [10, 12, 15, 20, 25, 31, 40, 45, 52, 60]

result = list(filter(lambda x:x % 10 == 0 or x % 10 == 5, numbers))

print(result)
```
