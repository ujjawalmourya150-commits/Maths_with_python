```python
# 1. Use map() + lambda to add 10 to every number.

numbers = [5, 12, 18, 25, 30]

result = list(map(lambda x: x + 10, numbers))

print(result)
```

```python
# 2. Use map() + lambda to multiply every number by 3.

numbers = [2, 4, 6, 8, 10]

result = list(map(lambda x: x * 3, numbers))

print(result)
```

```python
# 3. Use map() + lambda to get the last digit of every number.

numbers = [123, 456, 789, 1024, 555]

result = list(map(lambda x: x % 10, numbers))

print(result)
```

```python
# 4. Use map() + lambda to convert every negative number into a positive number.

numbers = [-5, 10, -20, 15, -8]

result = list(map(lambda x: abs(x), numbers))

print(result)
```

```python
# 5. Use map() + lambda to add corresponding elements of two lists.

numbers1 = [10, 20, 30, 40]
numbers2 = [1, 2, 3, 4]

result = list(map(lambda x, y: x + y, numbers1, numbers2))

print(result)
```

```python
# 6. Use map() + lambda to find the cube of every number.

numbers = [2, 3, 4, 5, 6]

result = list(map(lambda x: x ** 3, numbers))

print(result)
```

```python
# 7. Use map() + lambda to convert every name into uppercase.

names = ["ujjawal", "rahul", "aman", "rohit"]

result = list(map(lambda x: x.upper(), names))

print(result)
```

```python
# 8. Use map() + lambda to apply a 20% discount to every price.

prices = [500, 1000, 1500, 2000]

result = list(map(lambda x: x - (x * 20 / 100), prices))

print(result)
```

```python
# 9. Use map() + lambda to return:
# "Positive" if number > 0
# "Negative" if number < 0
# "Zero" if number == 0

numbers = [10, -5, 20, -8, 0, 15]

result = list(map(lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero", numbers))

print(result)
```

```python
# 10. Use map() + lambda to return:
# "Pass" if marks are 40 or above
# "Fail" if marks are below 40

marks = [85, 42, 67, 30, 90, 25]

result = list(map(lambda x: "Pass" if x >= 40 else "Fail", marks))

print(result)
```
