```python
# 1 . Write a lambda function that takes two numbers and returns their sum.

add = lambda a, b: a + b

result = add(10, 20)
print(result)
```

```python
# 2 . Write a lambda function that returns the square of a number.

square = lambda a:a**2

result = square(2)
print(result)
```

```python
# 3 . Write a lambda function that returns the cube of a number.

cube = lambda a:a**3

result = cube(2)
print(result)
```

```python
# 4 . Write a lambda function that returns "Even" if a number is even, otherwise "Odd".

even_odd = lambda x:"Even" if x%2==0 else "Odd"

result = even_odd(7)
print(result)
```

```python
# 5 . Write a lambda function that takes two numbers and returns the larger number.

largest  = lambda x,y:f"{x} is largest" if x>y else f"{y} is largest"
result = largest(3,4)
print(result)
```

```python
# 6 . Write a lambda function that takes two numbers and returns the smaller number.

smaller  = lambda x,y:f"{x} is smaller" if x<y else f"{y} is smaller"

result = smaller(3,4)
print(result)
```

```python
# 7 . Write a lambda function to convert Celsius into Fahrenheit.

fehrenheit = lambda c: c * 9/5 + 32

result = fehrenheit(31)
print(result)
```

```python
# 8 . Write a lambda function that returns the last digit of a number.

last_digit = lambda x:x%10

result = last_digit(12345)
print(result)
```

```python
# 9  . Write a lambda function that takes two numbers and returns their remainder.

remainder = lambda x,y:x%y

result = remainder(17,5)
print(result)
```

```python
# 10 . Check positive, negative, or zero

check = lambda x: f"{x} is a positive number" if x > 0 else (f"{x} is a negative number" if x < 0 else f"{x} is zero")

print(check(5))
print(check(-3))
print(check(0))
```
