```python
# 1. Use filter() + lambda to get numbers greater than 10.

numbers = [5, 12, 8, 20, 3, 15]

result = list(filter(lambda x:x>10,numbers))

print(result)
```

```python
# 2 . Use filter() + lambda to get only even numbers.

numbers = [10, 25, 15, 40, 8, 32, 19]

result = list(filter(lambda x:x%2==0,numbers))

print(result)
```

```python
# 3 . Use filter() + lambda to get numbers divisible by 3.

numbers = [6, 10, 15, 22, 27, 31, 36]

result = list(filter(lambda x:x%3==0,numbers))

print(result)
```

```python
# 4 . Use filter() + lambda to get only positive numbers.

numbers = [-5, 10, -2, 8, 0, -7, 15]

result = list(filter(lambda x:x>0,numbers))

print(result)
```

```python
# 5 .Use filter() + lambda to get numbers between 10 and 50, including 10 and 50.

numbers = [5, 12, 25, 7, 49, 51, 33, 60]

result = list(filter(lambda x:50>x>10,numbers))

print(result)
```

```python
# 6 . Use filter() + lambda to get numbers that are both even and greater than 20.

numbers = [12, 24, 15, 30, 7, 42, 19, 50]

result = list(filter(lambda x:x>20 and x%2==0,numbers))

print(result)
```

```python
# 7 .Use filter() + lambda to  get numbers divisible by both 3 and 5.

numbers = [10, 15, 20, 30, 45, 50, 60, 17]

result = list(filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers))

print(result)
```

```python
# 8 . Use filter() + lambda to  get words having more than 5 characters.

words = ["cat", "python", "apple", "AI", "machine", "data", "learning"]

result = list(filter(lambda x: len(x) > 5, words))

print(result)
```

```python
# 9 . Use filter() + lambda to  get names that start with "A".

names = ["Aman", "Rahul", "Ankit", "Priya", "Aditya", "Rohit"]

result = list(filter(lambda x: x.startswith("A"), names))

print(result)
```

```python
# 10 . Use filter() + lambda to get numbers whose last digit is 5.

numbers = [12, 25, 35, 47, 55, 68, 75, 81]

result = list(filter(lambda x: x % 10 == 5, numbers))

print(result)
```
