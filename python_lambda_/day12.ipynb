```python
# 1 . Write a lambda function that returns double the given number.

double = lambda x:x*2

print(double(2))
```

```python
# 2 . Write a lambda function that returns the absolute value of a number.

absolute = lambda x: x if x >= 0 else -x

print(absolute(-10))
```

```python
# 3 . Write a lambda function that returns True if a number is divisible by 5, otherwise False.

divi = lambda x:f"{x} is divisible by 5" if x%5==0 else f"{x} is not divisible by 5"

print(divi(43))
```

```python
# 4. Find the largest of three numbers

largest = lambda a,b,c:f"{a} is largest" if b<a>c else(f"{b} is largest" if a<b>c else f"{c} is the largest")

print(largest(23,43,12))
```

```python
# 5 . Write a lambda function that takes three numbers and returns their sum.

total = lambda a,b,c: f" total sum of three numbers are {a+b+c}"

print(total(2,3,5))
```

```python
# 6 . Write a lambda function that returns the first digit of a number.

first = lambda x:f" first digit of the number is {str(x)[0]}"

print(first(223))
```

```python
# 7 . Write a lambda function using:

si = lambda P,R,T: f"Simple intrest is {(P*R*T)/100}"

print(si(1000,5,2))
```

```python
# 8 . Write a lambda function that takes length and width and returns the area.

area = lambda l, b: f" Area of rectangle : {l * b}"

print(area(10, 5))
```

```python
# 9 . Write a lambda function that returns True if the number is greater then 100, otherwise False.

check = lambda x:True if x > 100 else False

print(check(101))
```

```python
# 10 . Write a lambda function that returns the last two digits of a number.

last = lambda x: f"The last two digit of the number are : {str(x)[-2:]}"

print(last(123))
```
