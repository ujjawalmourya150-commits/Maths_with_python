```python
# 1 . Use reduce() + lambda to find the sum of all numbers.

from functools import reduce

numbers = [10, 20, 30, 40, 50]

result = reduce(lambda x,y:x+y,numbers)

print(result)
```

```python
# 2. Use reduce() + lambda to find the product of all numbers.

from functools import reduce

numbers = [2, 3, 4, 5]

result = reduce(lambda x,y:x*y,numbers)

print(result)
```

```python
# 3. Use reduce() + lambda to find the largest number.

from functools import reduce

numbers = [12, 45, 23, 67, 34, 89, 10]

result = reduce(lambda x,y: x if x>y else y ,numbers)

print(result)
```

```python
# 4. Use reduce() + lambda to find the smallest number.

from functools import reduce

numbers = [45, 12, 78, 3, 56, 9]

result = reduce(lambda x,y: x if x<y else y ,numbers)

print(result)
```

```python
# 5. Use reduce() + lambda to find the total price.

prices = [500, 1200, 350, 800, 150]

result = reduce(lambda x,y:x+y,prices)

print(result)
```

```python
# 6. Use reduce() + lambda to find the sum of only the even numbers.

from functools import reduce

numbers = [10, 15, 20, 25, 30, 35, 40]

result = reduce(lambda x, y: x + y, filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
# 7. Use reduce() + lambda to find the sum of only the odd numbers.

numbers = [11, 20, 31, 42, 53, 64, 75]

result = reduce(lambda x, y: x + y, filter(lambda x: x % 2 != 0, numbers))

print(result)
```

```python
# 8 .Use reduce() + lambda to find the total number of characters in all the words.

words = ["Python", "AI", "Machine", "Learning"]

result = reduce(lambda x, y: x + len(y), words, 0)

print(result)
```

```python
# 9. Use reduce() + lambda to find the longest word.

words = ["Python", "Artificial", "AI", "Machine", "Learning"]

result = reduce(lambda x, y: x if len(x) > len(y) else y, words)

print(result)
```

```python
# 10 . Use reduce() + lambda to find the shortest word.

words = ["Python", "AI", "Machine", "Code", "ML"]

result = reduce(lambda x, y: x if len(x) < len(y) else y, words)

print(result)
```
