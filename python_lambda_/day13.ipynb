```python
#  1 .Write a lambda function that returns 3 times a number.

Triple = lambda x:x*3

print(Triple(5))
```

```python
#  2 .Return True if a number is divisible by both 2 and 3.

Divi = lambda x:x%2==0 and x%3==0

print(Divi(6))
```

```python
# 3 . Return the square if the number is even, otherwise return the number itself.

Square_even = lambda x:x**2 if x%2==0 else x

print(Square_even(6))
```

```python
# 4 . Find the largest of two numbers

largest = lambda x,y:"x is largest " if x>y else "y is largest"

print(largest(2,3))
```

```python
# 5 . Calculate the perimeter of a rectangle 

perimeter = lambda l, b: 2 * (l + b)

print(perimeter(10, 5))
```

```python
# 6 . Calculate the area of a circle  

area = lambda r: 3.14 * r * r

print(area(5))
```

```python
# 7 . Return True if it is between 10 and 50, otherwise False.

check = lambda x: 10 <= x <= 50

print(check(25))
```

```python
# 8 . Convert minutes to seconds

seconds = lambda minutes: minutes * 60

print(seconds(5))
```

```python
# 9 .Take price and discount_percent as arguments.

discount = lambda price, discount_percent: price - (price * discount_percent / 100)

print(discount(1000, 20))
```

```python
# 10 . Take age as input and return "Eligible" if age is 18 or above, otherwise "Not Eligible".

eligible = lambda Age: "Eligible" if Age >= 18 else "Not eligible"
print(eligible(20))
```
