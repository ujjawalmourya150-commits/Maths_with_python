```python
# 1. Write a function that takes a list and returns a new list without duplicate values.

numbers = [1, 2, 2, 3, 1, 4, 3]

def without_duplicate(n):
    l = []

    for i in n:
        if i not in l:
            l.append(i)

    return l

print(without_duplicate(numbers))
```

```python
# 2. Write a function that counts how many times each number occurs.

def frequency(n):
    visited = []

    for i in n:
        if i not in visited:
            count = 0

            for j in n:
                if i == j:
                    count += 1

            print(i, "occurs", count, "times")
            visited.append(i)


frequency([2, 3, 2, 5, 3, 2])
```

```python
# 3 . Write a function frequency(numbers, target) that returns how many times target occurs.

def frequency(numbers, target):
    count = 0

    for i in numbers:
        if i == target:
            count += 1

    return count


f = frequency([2, 3, 2, 5, 3, 2], 2)
print(f"it occurs {f} times")
```

```python
# 4 .  Write a function unique_values(numbers) that returns a list containing only values that occur once.

def unique_values(numbers):
    result = []

    for i in numbers:
        count = 0

        for j in numbers:
            if i == j:
                count += 1

        if count == 1:
            result.append(i)

    return result


u = unique_values([2, 3, 2, 5, 3, 7])
print(u)
```

```python
# 5 .  Write a function find_missing(numbers) where the list contains numbers from 1 to n with one number missing.

def find_missing(numbers):
    n = len(numbers) + 1

    for i in range(1, n + 1):
        if i not in numbers:
            return i


m = find_missing([1, 2, 3, 5, 6])
print(f" The missing numebr is {m}")
```

```python
# 6 . Write a function separate_numbers(numbers) that creates two lists: even numbers , odd numbers

def separate_numbers(numbers):

    Odd = []
    Even = []

    for i in numbers:
        if i % 2 == 0:
            Even.append(i)
        else:
            Odd.append(i)

    return Odd,Even
    
s = separate_numbers([2,3,7,4,8,9,14])
print(s)
```

```python
# 7 .  Write a function largest_difference(numbers) that finds the difference between the largest and smallest number.

def largest_difference(numbers):

    largest = numbers[0]
    smallest = numbers[0]
    difference = 0

    for i in numbers:
        if largest < i:
            largest = i

        if smallest > i:
            smallest = i

    difference = largest - smallest
    return difference

d = largest_difference([4,2,1,6,34,7,45,3,12,])
print(f"The largest difference is {d}")
```

```python
# 8 . Write a function count_digits(n) using a while loop.

def count_digits(n):

    count = 0

    while n > 0:

        n = n//10
        count += 1

    return count

c = count_digits(12345678)
print(c)
```

```python
# 9 . Write a function reverse_number(n) using a while loop.

def reverse_number(n):

    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    return reverse

r = reverse_number(123)
print(r)
```

```python
# 10 . Write a function digit_sum(n) using a while loop.

def digit_sum(n):

    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10
    return total

d = digit_sum(123)
print(d)
```
