```python
# 4 . Function for prime number 

def Prime(n1):
    count = 0
    for i in range(1,n1+1):
        if n1%i==0:
            count+=1
    if count == 2:
        print("Prime number")
    else:
        print("Not prime number")
Prime(7)
```

```python
# 4 . Function for prime number (Different logic)

def Prime(n1):
    for i in range(2,n1):
        if n1 % i == 0:
            return False
    else:
            return True
Prime(21)
```

```python
# 5 . Write a function square(n) that returns the square of a number.

num = int(input("Enter a number : "))

def square(n):
    return n*n

s = square(num)

print(f"Square of {num} is {s}")
```

```python
#  6  .Write a function factorial(n) that returns the factorial of a number.

num = int(input("Enter a number : "))

def Fact(n):
    factorial = 1
    for i in range(1,n+1):
        factorial = factorial * i
    return factorial
f = Fact(num)
print(f" Factorial of {num} is {f}")
```

```python
# 7 . Write a function count_vowels(word) that takes a string and returns the number of vowels.

def count_vowels(word):
    count = 0
    vowels = ['a', 'e', 'i', 'o', 'u']

    for i in word:
        if i in vowels:
            count += 1

    return count

s = count_vowels("ujjawal")
print(s)
```

```python
# 8 . Write a function second_largest(numbers) that takes a list and returns the second-largest number.
l = [2,3,5,1,9,7,8]

def second_largest(numbers):

    largest = l[0]
    second = l[0]

    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i
    return second

s = second_largest(l)
print(f"The second largest number from list is {s}")  
```

```python
# 9 . Write a function count_digits(n) that takes an integer and returns the number of digits.

def count_digits(n):
    count = 0

    for i in str(n):
        count += 1

    return count

print(count_digits(123))
    
```

```python
# 10 . Write a function reverse_number(n) that takes a number and returns its reverse.

def reverse_number(n):
    reverse = 0

    for i in range(len(str(n))):
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    return reverse

print(reverse_number(12345))
```
