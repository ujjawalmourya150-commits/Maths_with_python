```python
# 1 . Write a function is_palindrome(n) that returns True if the number is a palindrome, otherwise False.

def is_palindrome(n):
    original = n
    reverse = 0
    for i in range(len(str(n))):
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    if reverse == original:
        return True
    else:
        return False
    
s = is_palindrome(121)
print(s)
```

```python
# 2 . Write a program to find the second smallest

num = [10,45,23,87,67,89,34]

def second_smallest(l):
    smallest = l[0]
    second = l[1]

    for i in l:
        if i < smallest:
            second = smallest
            smallest = i

        elif i < second and i != smallest:
            second = i

    print("smallest number :", smallest)
    print("second smallest number :", second)

s = second_smallest(num)
print(s)
```

```python
# 3 . Write a function digit_sum(n) that returns the sum of all digits.

def digit_sum(n):
    add = 0

    for i in str(n):
        add = add + int(i)
    return add

s = digit_sum(12345)
print(s)
```
