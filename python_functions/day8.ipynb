```python
# 1 . Write a function is_palindrome(n) that returns True if the number is a palindrome, otherwise False.

def is_palindrome(n):
    original = n
    reverse = 0
    for i in range(len(str(n))):
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    if reverse == original:
        return True
    else:
        return False
    
s = is_palindrome(121)
print(s)
```

```python
# 2 . Write a program to find the second smallest

num = [10,45,23,87,67,89,34]

def second_smallest(l):
    smallest = l[0]
    second = l[1]

    for i in l:
        if i < smallest:
            second = smallest
            smallest = i

        elif i < second and i != smallest:
            second = i

    print("smallest number :", smallest)
    print("second smallest number :", second)

s = second_smallest(num)
print(s)
```

```python
# 3 . Write a function digit_sum(n) that returns the sum of all digits.

def digit_sum(n):
    add = 0

    for i in str(n):
        add = add + int(i)
    return add

s = digit_sum(12345)
print(s)
```

```python
# 4 . Write a function find_max(numbers) that takes a list and returns the largest number without using max().
l = [4 , 7 , 100 , 2 , 9 , 4 , 56 , 1]

def find_max(numbers):
    maximum = numbers[0]
    for i in numbers:
        if maximum < i:
            maximum = i
    return maximum

f = find_max(l)
print(f)
```

```python
# 5 .Write a function find_min(numbers) that returns the smallest number without using min().

l = [4 , 7 , 100 , 2 , 9 , 4 , 56 , 1 , 0]

def find_min(numbers):
    minimum = numbers[0]
    for i in numbers:
        if minimum > i:
            minimum = i
    return minimum

f = find_min(l)
print(f)
```

```python
# 6 . Write a function count_even(numbers) that returns how many even numbers are present.

l = [2,4,3,7,9,22,45,6,78,99,23]

def count_even(numbers):
    count = 0
    for i in numbers:
        if i % 2 ==0:
            count += 1
    return count

c = count_even(l)
print(f"Even numbers count is {c}")        
```

```python
# 8 .  Write a function common_elements(a, b) that returns the elements present in both lists.

l1 = [1,2,3,4] 
l2 = [3,4,5,6]

def common_elements(a, b):

    common = []

    for i in a:
        for j in b:
            if i == j:
                common.append(i)
    return common

c = common_elements(l1,l2)
print(f"Common elements list is : {c}")
```

```python
# 9. Return the third-largest number without using sort().

numbers = [34,23,45,67,2,87,89]

def third_largest(n):
    largest = n[0]
    second = n[0]
    third = n[0]

    for i in n:
        if largest < i:
            third = second
            second = largest
            largest = i
        elif second < i and i != largest:
            third = second
            second = i
        elif third < i and i != second and i != largest:
            third = i

    return third

t =third_largest(numbers)
print(f"Third largest number from list : {t}")
```

```python
# 10 . Write a function that checks whether a number is an Armstrong number.

def Armstrong_number(n):
    original = n
    power = len(str(n))
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit ** power
        n = n // 10

    if total == original:
        return True
    else:
        return False

print(Armstrong_number(int(input("Enter a number to check number is armstrong number or not : "))))
```
