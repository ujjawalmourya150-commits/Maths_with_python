```python
# 1 . Write a function digit_product(n) that multiplies all digits.

def digit_product(n):

    Product = 1

    while n > 0:

        digit = n % 10
        Product = Product * digit
        n = n//10

    return Product

d = digit_product(123)
print(d)
```

```python
# 2 . Write a function count_even_digits(n). 

def count_even_digits(n):

    count = 0

    while n > 0:
        digit = n % 10

        if digit % 2 == 0:
            count += 1

        n = n // 10

    return count

c = count_even_digits(123456)
print(c)
```

```python
# 3 . Write a function largest_digit(n) without converting the number to a string.

def largest_digit(n):

    largest = 0

    while n > 0:
        digit = n % 10

        if digit > largest:
            largest = digit

        n = n // 10

    return largest

l = largest_digit(58321)
print(l)
```

```python
# 5 . Write a function is_palindrome(n) using a while loop.

def is_armstrong(n):

    original = n
    power = len(str(n))
    product = 0

    while n > 0:

        digit = n % 10
        product = product + digit ** power
        n = n // 10

    if original == product:
        return True
    else:
        return False

a = is_armstrong(153)
print(a)
```

```python
# 7 . Write a function product_of_digits(n) using only a while loop.

def product_of_digits(n):
    product = 1
    while n > 0:
        digit = n % 10
        product = product * digit
        n = n//10
    return product

p = product_of_digits(123)
print(p)
```

```python
# 8 . Write a function that returns the first repeated number in a list.

def repeated_number(n):

    seen = []
    i = 0

    while i < len(n):

        if n[i] in seen:
            return n[i]

        seen.append(n[i])
        i += 1

    return None

r = repeated_number([1,3,7,1,4,2,5,2])
print(r)
```

```python
# 9 . Write a function that returns the first non-repeated number.

def non_repeated(numbers):
    i = 0

    while i < len(numbers):
        count = 0
        j = 0

        while j < len(numbers):
            if numbers[i] == numbers[j]:
                count += 1
            j += 1

        if count == 1:
            return numbers[i]

        i += 1

    return None


numbers = [4, 7, 2, 7, 4, 9, 2]
print(non_repeated(numbers))
```
