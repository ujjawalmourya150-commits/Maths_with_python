```python
# DAY 35 - QUESTION 1
#
# Given a list of numbers:
# 1. Find the frequency of every distinct number.
# 2. Print only numbers appearing MORE THAN TWICE.
# 3. Print the number and its frequency.
# 4. Print each number only once.
# 5. Preserve first-appearance order.
#
# Do NOT use count(), set(), Counter(), or dictionaries.

numbers = [5, 2, 5, 3, 2, 5, 7, 3, 2, 3, 3, 8, 8, 8, 8]
l=[]
more_than_twice=[]
for i in range(len(numbers)):
    count=0
    
    if numbers[i] not in l:
        l.append(numbers[i])

        for j in range(len(numbers)):
            if numbers[i]==numbers[j]:
                count+=1
        

        if count > 2:
            more_than_twice.append(numbers[i])
            print(numbers[i],"appears",count,"times")

            
print(f"\n Appears more than twice : {more_than_twice}")
```

```python
# DAY 35 - QUESTION 2
#
# Given two lists:
# 1. Find numbers appearing in BOTH lists.
# 2. Their frequency in list a must be GREATER
#    than their frequency in list b.
# 3. Print the number and both frequencies.
# 4. Print each number only once.
# 5. Preserve first-appearance order from list a.
#
# Do NOT use count(), set(), Counter(), or dictionaries.


a = [4, 2, 4, 7, 2, 4, 9, 5, 5, 8]
b = [4, 2, 7, 7, 9, 5, 8, 8, 8]

# Find numbers appearing in BOTH lists.
l=[]
for i in a:
    if i not in l:
        if i in b:
            l.append(i)

print("Numbers appearing in both lists : ",l)

# Calculate frequencies

for i in l:
    freq_a = 0
    freq_b = 0

    for j in a:
        if i == j:
            freq_a += 1

    for j in b:
        if i == j:
            freq_b += 1

    if freq_a > freq_b:
        print(i, "appears", freq_a, "times in a and", freq_b, "times in b")
```

```python
# DAY 35 - QUESTION 4
#
# Given two lists:
# 1. Find common numbers.
# 2. Their frequencies must be DIFFERENT in both lists.
# 3. Print numbers whose frequency in a is LESS
#    than their frequency in b.
# 4. Print each number only once.
# 5. Preserve first-appearance order from list a.
#
# Do NOT use count(), set(), Counter(), or dictionaries.

a = [6, 3, 6, 2, 3, 6, 8, 2, 5, 5]
b = [6, 3, 3, 3, 2, 8, 8, 8, 5, 9]
l=[]
less=[]
for i in a:
    if i not in l:
        if i in b:
            freq_a=0
            freq_b=0
            for j in a:
                if i==j:
                    freq_a+=1
            for k in b:
                if i==k:
                    freq_b+=1

            if freq_b != freq_a:
                if i not in less:
                    if freq_a < freq_b:
                        less.append(i)
        l.append(i)
print("Numbers whose frequency in a is less than b:", less)
```

```python
# DAY 35 - QUESTION 6
#
# Given a list of numbers:
# 1. Find the frequency of every distinct number.
# 2. Calculate the average frequency.
# 3. Print numbers having frequency GREATER
#    than the average frequency.
# 4. Print their frequencies.
# 5. Preserve first-appearance order.
#
# Do NOT use count(), sum(), set(), Counter(),
# or dictionaries.

numbers = [2, 4, 2, 7, 4, 2, 9, 9, 9, 5, 5, 5, 5, 8]
l=[]
freq=[]
total=0
print("Numbers and their frequencies : ")
for i in numbers:
    count=0
    if i not in l:
        l.append(i)
        for j in numbers:
            if i==j:
                count+=1
        freq.append(count)
        print(i ," appears ",count)

for f in freq:
    total+=f
avg=total/len(freq)

print("\nAvearge : ", round(avg,2))

print("\nNumbers having frequency GREATER than the average frequency :")
for i in range(len(freq)):
    if freq[i]>avg:
        print(l[i]," appear " , freq[i])
```

```python
# DAY 35 - QUESTION 7
#
# Given two lists:
# 1. Find numbers appearing in BOTH lists.
# 2. Calculate the frequency difference:
#    frequency in a - frequency in b.
# 3. Print numbers whose frequency difference
#    is GREATER THAN 1.
# 4. Print the number and frequency difference.
# 5. Print each number only once.
# 6. Preserve first-appearance order from list a.
#
# Do NOT use count(), set(), Counter(), or dictionaries.

a = [3, 5, 3, 2, 5, 3, 7, 2, 8, 5, 9]
b = [3, 5, 2, 5, 8, 9, 9]
l=[]
greater_than_1=[]
for i in a:
    if i not in l:
        if i in b:
            l.append(i)
for u in l:
    freq_a=0
    freq_b=0
    for j in a:
        if u==j:
            freq_a+=1
    for k in b:
        if u==k:
            freq_b+=1

    difference = freq_a - freq_b
    print(f"The frequency difference of {u} in a and b is {difference}")

    if difference>1:
        greater_than_1.append(u)
print("\n Numbers whose frequency difference is GREATER THAN 1 :",greater_than_1)

            
```

```python
# DAY 35 - QUESTION 9
#
# Given a list of numbers:
# 1. Find the frequency of every distinct number.
# 2. Find the highest and lowest frequencies.
# 3. Calculate the difference between them.
# 4. Print numbers having the highest frequency.
# 5. Print numbers having the lowest frequency.
# 6. Preserve first-appearance order.
#
# Do NOT use count(), max(), min(), sorted(),
# set(), Counter(), or dictionaries.

numbers = [6, 3, 6, 2, 3, 6, 8, 2, 3, 3, 5, 5, 9, 9, 9, 9]
l=[]
highest=0
highest_num=[]
lowest_num=[]

lowest=None
print("Frequency of every distinct number :")
for i in numbers:
    count=0
    if i not in l:
        for j in numbers:
            if i==j:
                count+=1
        print(i ," appears ",count)
        l.append(i)

        # Finding highest frequency
        if count > highest:
            highest = count
            highest_num = [i]

        elif count == highest:
            highest_num.append(i)

        # Finding lowest frequency
        if lowest == None or count < lowest:
            lowest = count
            lowest_num = [i]

        elif count == lowest:
            lowest_num.append(i)
            

difference = highest-lowest

print("\nHighest frequency:", highest)
print("Lowest frequency:", lowest)

print("\nDifference between highest frequency and lowest frequency : ",difference)

print("\nNumbers with highest frequency:", highest_num)
print("Numbers with lowest frequency:", lowest_num)
```

```python
# DAY 35 - QUESTION 10 (FINAL CHALLENGE)
#
# Given a list of numbers:
# 1. Find the frequency of every distinct number.
# 2. Calculate the average frequency.
# 3. Find the SECOND-HIGHEST DISTINCT frequency.
# 4. Find the LOWEST frequency.
# 5. Print numbers satisfying BOTH conditions:
#       a. Frequency is greater than average.
#       b. Frequency is equal to the second-highest
#          distinct frequency.
# 6. Also print numbers having the lowest frequency.
# 7. Print each number only once.
# 8. Preserve first-appearance order.
# 9. Handle cases where fewer than two distinct
#    frequencies exist.
#
# Do NOT use count(), sum(), max(), min(), sorted(),
# set(), Counter(), or dictionaries.


numbers = [6, 3, 6, 2, 3, 6, 8, 2, 3, 3, 5, 5, 5, 9, 9, 9, 9]

l = []
freq = []

total = 0
highest = 0
second = None
lowest = None

result = []
lowest_num = []

print("Frequency of every distinct number:")

# STEP 1: Calculate frequency of every distinct number

for i in numbers:
    count = 0

    if i not in l:
        for j in numbers:
            if i == j:
                count += 1

        print(i, "appears", count)

        l.append(i)
        freq.append(count)

        total += count

        # Finding highest and second-highest
        if count > highest:
            second = highest if highest != 0 else second
            highest = count

        elif count < highest:
            if second == None or count > second:
                second = count

        # Finding lowest frequency
        if lowest == None or count < lowest:
            lowest = count

# STEP 2: Calculate average frequency

avg = total / len(freq)

# STEP 3: Handle fewer than two distinct frequencies
if second == None:
    print("\nSecond-highest distinct frequency does not exist.")

else:
    print("\nAverage frequency:", avg)
    print("Highest frequency:", highest)
    print("Second-highest distinct frequency:", second)
    print("Lowest frequency:", lowest)

    # STEP 4: Find numbers satisfying both conditions

    for i in range(len(l)):
        if freq[i] > avg and freq[i] == second:
            if l[i] not in result:
                result.append(l[i])

    # STEP 5: Find numbers having lowest frequency

    for i in range(len(l)):
        if freq[i] == lowest:
            if l[i] not in lowest_num:
                lowest_num.append(l[i])

    print("\nNumbers satisfying both conditions:", result)
    print("Numbers having lowest frequency:", lowest_num)

```
