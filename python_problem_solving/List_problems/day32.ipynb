```python
# DAY 32 - QUESTION 1
#
# Move all zeros to the end of the list.
# Preserve the original order of non-zero numbers.
# Create a new list.
#
# Do NOT use sorted(), set(), or dictionaries.

numbers = [0, 4, 0, 2, 7, 0, 3, 2]

new_list = []

for i in numbers:
    if i != 0:
        new_list.append(i)

for i in numbers:
    if i == 0:
        new_list.append(i)

print(new_list)
```

```python
# DAY 32 - QUESTION 2
#
# 1. Find numbers that appear in BOTH lists.
# 2. Print each number only once.
# 3. Preserve first-appearance order from list a.
#
# Do NOT use set(), dictionaries, or built-in sorting.

a = [4, 2, 7, 4, 9, 2, 5]
b = [2, 8, 4, 2, 5, 10]

common=[]

for i in a:
    for j in b:
            if i==j and i not in common:
                common.append(i)

print(common)
            
```

```python
# DAY 32 - QUESTION 5
#
# 1. Calculate the average of all numbers.
# 2. Print distinct numbers strictly above the average.
# 3. Preserve first-appearance order.
#
# Do NOT use sum(), set(), or dictionaries.

numbers = [10, 20, 30, 20, 40, 50, 10, 60]

total=0

for i in numbers:
    total+=i
avg=total/len(numbers)
print(avg)

above_avg=[]
for i in numbers:
    if i>avg and i not in above_avg :
        above_avg.append(i)
print(above_avg)
```

```python
# DAY 32 - QUESTION 7
#
# 1. Find all pairs of numbers whose sum equals target.
# 2. Use different positions for each pair.
# 3. Do not print the same pair twice.
# 4. Handle duplicate numbers correctly.
#
# Do NOT use set(), dictionaries, or sorted().

numbers = [2, 7, 4, 5, 3, 6, 2]
target = 9

pairs = []

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] + numbers[j] == target:
            a = min(numbers[i], numbers[j])
            b = max(numbers[i], numbers[j])

            pair = [a, b]

            if pair not in pairs:
                pairs.append(pair)
                print(pair)
```

```python
# DAY 32 - QUESTION 8
#
# 1. Count the frequency of each distinct number.
# 2. Create three lists:
#       once   -> numbers appearing exactly once
#       twice  -> numbers appearing exactly twice
#       repeat -> numbers appearing more than twice
# 3. Print each number only once in its category.
# 4. Preserve first-appearance order.
#
# Do NOT use .count(), Counter(), set(), or dictionaries.

numbers = [3, 5, 3, 2, 5, 3, 7, 2, 8, 5]

l=[]
freq=[]

for i in numbers:
    if i not in l:
        l.append(i)
        freq.append(1)
    else:
        index=l.index(i)
        freq[index]+=1

print("Frequencies :",freq)


once=[]
twice=[]
repeat=[]

for index in range(len(l)):
    if freq[index]==1:
        once.append(l[index])
    elif freq[index]==2:
        twice.append(l[index])
    else:
        repeat.append(l[index])

print("\nOnce:",once)
print("Twice:",twice)
print("Repeated:",repeat)
```
