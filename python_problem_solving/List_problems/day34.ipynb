```python
# DAY 34 - QUESTION 1
#
# Given a list of numbers:
# 1. Find numbers that appear MORE THAN TWICE.
# 2. Print each number only once.
# 3. Preserve first-appearance order.
#
numbers = [4, 2, 4, 7, 4, 2, 9, 2, 2, 5, 7, 7, 7]

new_list=[]
result=[]

for i in range(len(numbers)):

    if numbers[i] not in new_list:
        new_list.append(numbers[i])

        count=0

        for j in range(len(numbers)):
            if numbers[i]==numbers[j]:
                count+=1

    if count>2 and numbers[i] not in result:
        result.append(numbers[i])
print(result)
        
```

```python
# DAY 34 - QUESTION 5
#
# Given a list:
# 1. Separate consecutive numbers into groups.
# 2. Each group should contain identical consecutive numbers.
# 3. Print the groups as a new nested list.

numbers = [4, 4, 4, 2, 2, 7, 7, 1, 4, 4, 9, 9, 9]

l=[]
temp=[numbers[0]]

for i in range(0,len(numbers)):
    if numbers[i]==numbers[i-1]:
        temp.append(numbers[i])
    else:
        l.append(temp)
        temp=[numbers[i]]

l.append(temp)

print("Consecutive numbers into groups are : ",l)
```

```python
# DAY 34 - QUESTION 6
#
# Find the LONGEST CONTIGUOUS strictly decreasing sequence.
# 1. Print the sequence.
# 2. Print its length.
# 3. If multiple sequences have the same length,
#    print the FIRST such sequence.
#
numbers = [12, 10, 8, 15, 13, 11, 9, 7, 14, 6, 4]
longest=[]
temp=[numbers[0]]

for i in range(len(numbers)-1):
    if numbers[i]>numbers[i+1]:
        temp.append(numbers[i+1])
    else:
        if len(temp)>len(longest):
            longest=temp
        temp=[numbers[i+1]]
if len(temp)>len(longest):
    longest=temp

print("Longest sequence:", longest)
print("Length:", len(longest))
```

```python
# DAY 34 - QUESTION 7
#
# Given a list and a target:
# 1. Find all unique pairs whose sum equals target.
# 2. Each pair must use two different positions.
# 3. Do not print the same pair more than once.
# 4. Preserve the order in which pairs are first found.
#
# Do not use sorted(), set(), or dictionaries.
#
numbers = [2, 4, 3, 5, 7, 2, 6, 4, 8, 1]
target = 9
Unique_pairs=[]
for i in range(len(numbers)):
    for j in range(i+1,len(numbers)):

        if numbers[i] + numbers[j] == target:
            pair = [numbers[i], numbers[j]]

            if pair not in Unique_pairs and pair[::-1] not in Unique_pairs:
                Unique_pairs.append(pair)

print("Unique Pairs that sum is target value are : ",Unique_pairs)
```

```python
# DAY 34 - QUESTION 8
#
# Given a list:
# 1. Find the number(s) having the HIGHEST frequency.
# 2. If multiple numbers share the highest frequency,
#    print all of them.
# 3. Print each number only once.
# 4. Preserve first-appearance order.
#
numbers = [4, 2, 4, 7, 2, 4, 9, 7, 2, 5, 5, 5, 8, 8]

highest_freq = 0
result = []

for i in range(len(numbers)):
    count = 0

    for j in range(len(numbers)):
        if numbers[i] == numbers[j]:
            count += 1

    if count > highest_freq:
        highest_freq = count
        result = [numbers[i]]

    elif count == highest_freq:
        if numbers[i] not in result:
            result.append(numbers[i])

print("Highest frequency:", highest_freq)
print("Numbers:", result)
```

```python

# DAY 34 - QUESTION 9
#
# Given two lists:
# 1. Find numbers appearing in BOTH lists.
# 2. Their frequencies must be DIFFERENT in the two lists.
# 3. Print the number and both frequencies.
# 4. Print each number only once.
# 5. Preserve first-appearance order from list a.
#
a = [3, 5, 3, 2, 5, 3, 7, 2, 8, 5, 9]
b = [3, 5, 2, 5, 8, 8, 9, 9, 9]

l=[]

for i in a:
    if i in b and i not in l:
        l.append(i)

for i in l:
    freq_a=0
    freq_b=0

    for j in a:
        if i==j:
            freq_a+=1

    for j in b:
        if i==j:
            freq_b+=1

    if freq_a!=freq_b:
        print(i, "Frequency in a:", freq_a, "Frequency in b:", freq_b)
```
