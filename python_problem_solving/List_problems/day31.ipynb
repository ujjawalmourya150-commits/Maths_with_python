```python
# DAY 31 - QUESTION 1
#
# Calculate the total and average of all numbers.
# Do NOT use sum().

numbers = [12, 25, 18, 30, 15, 40, 20]

total=0
for i in numbers:
    total+=i
    
avg=total/len(numbers)
print(f"Total:{total}")
print(f"\nAverage:{avg}")
```

```python
# DAY 31 - QUESTION 2
#
# Count how many even and odd numbers are present.

numbers = [12, 7, 24, 15, 9, 32, 18, 5, 40]
even=0
odd=0
for i in numbers:
    if i%2==0:
        even+=1
    else:
        odd+=1
print(f"Total even numbers : {even}")
print(f"Total odd numbers : {odd}")
```

```python
# DAY 31 - QUESTION 3
#
# Find the SECOND-HIGHEST DISTINCT number.
# Do NOT use max(), sorted(), or set().

numbers = [45, 20, 45, 70, 35, 70, 60, 50]

highest=0
second=0

for i in numbers:
    if i>highest:
        second=highest
        highest=i
    elif second<i and highest>i:
        second=i
print(f"The second highest number {second}")
```

```python
# DAY 31 - QUESTION 4
#
# Remove duplicate numbers while preserving their original order.
# Create a new list.
# Do NOT use set().

l=[]

numbers = [4, 2, 4, 7, 2, 9, 7, 5, 9]

for i in numbers:
    if i not in l:
        l.append(i)

print(l)
```

```python
# DAY 31 - QUESTION 5
#
# Separate positive and negative numbers into two lists.
# Count zeros separately.

numbers = [12, -5, 0, 8, -3, 0, 15, -10, 7]

positive=[]
negative=[]
zero=[]
for i in numbers:
    if i>0:
        positive.append(i)
    elif i==0:
        zero.append(i)
    else:
        negative.append(i)

print("Positive : ",positive)
print("Negative : ",negative)
print("Zero : ",zero)

```

```python
# DAY 31 - QUESTION 6
#
# 1. Count the frequency of each distinct number.
# 2. Print each distinct number and its frequency.
# 3. Preserve the order of first appearance.
# Do NOT use .count() or Counter().

numbers = [4, 2, 4, 7, 2, 4, 9, 7, 2, 5]

checked = []

for i in numbers:
    if i not in checked:
        count = 0

        for j in numbers:
            if i == j:
                count += 1

        print(i, "appears", count, "times")
        checked.append(i)

```

```python
# DAY 31 - QUESTION 7
#
# Find the THIRD-HIGHEST DISTINCT number.
# Handle the case where fewer than three distinct numbers exist.
# Do NOT use max(), sorted(), or set().


numbers = [30, 50, 20, 50, 70, 40, 70, 60, 30]

highest = None
second = None
third = None

for i in numbers:
    if highest is None or i > highest:
        if i != highest:
            third = second
            second = highest
            highest = i

    elif i != highest and (second is None or i > second):
        third = second
        second = i

    elif i != highest and i != second and (third is None or i > third):
        third = i

if third is not None:
    print("Third-highest distinct number:", third)
else:
    print("Fewer than three distinct numbers")
```

```python
# DAY 31 - QUESTION 8
#
# Group numbers into three categories:
# High   -> number >= 50
# Medium -> number >= 20 and < 50
# Low    -> number < 20
# Create three separate lists.

numbers = [15, 60, 35, 80, 10, 45, 55, 20, 5, 70]
high=[]
medium=[]
low=[]

for i in numbers:
    if i>=50:
        high.append(i)
    elif i>=20 and i<50:
        medium.append(i)
    else:
        low.append(i)
print("high :")
print(high)
print("medium :")
print(medium)
print("low :")
print(low)
```

```python
# DAY 31 - QUESTION 9
#
# 1. Calculate the average of all numbers.
# 2. Print numbers strictly above the average.
# 3. Preserve their original order.
# Do NOT use sum().

numbers = [10, 20, 30, 40, 50, 60, 70]

total=0

for i in numbers:
    total+=i
    
avg=total/len(numbers)

new_list=[]

for i in numbers:
    if i>avg:
        new_list.append(i)

print(new_list)

```
