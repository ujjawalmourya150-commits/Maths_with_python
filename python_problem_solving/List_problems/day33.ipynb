```python
# DAY 33 - QUESTION 2
#
# Merge two already-sorted lists into one sorted list.
# Do NOT use sorted() or sort().
# Do NOT use dictionaries or sets.
#
# Hint: Use two indexes and compare elements.
a = [2, 5, 8, 12, 17]
b = [1, 4, 6, 10, 15, 20]

sorted_list = []

i = 0
j = 0

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        sorted_list.append(a[i])
        i += 1
    else:
        sorted_list.append(b[j])
        j += 1

while i < len(a):  # runs if b finishes first
    sorted_list.append(a[i])
    i += 1

while j < len(b):  # runs if a finishes first
    sorted_list.append(b[j])
    j += 1

print(sorted_list)
```

```python
# DAY 33 - QUESTION 3
#
# Find numbers that appear EXACTLY ONCE in BOTH lists.
# Print each qualifying number only once.
# Preserve first-appearance order from list a.
#
# Do NOT use .count(), set(), Counter(), or dictionaries.

a = [4, 2, 7, 4, 9, 2, 5, 11]
b = [2, 8, 4, 2, 5, 10, 9, 9, 11]

new_list = []

for i in a:
    count_a = 0
    count_b = 0

    for j in a:
        if i == j:
            count_a += 1

    for j in b:
        if i == j:
            count_b += 1

    if count_a == 1 and count_b == 1:
        if i not in new_list:
            new_list.append(i)

print(new_list)
```

```python
# DAY 33 - QUESTION 4
#
# Find the LONGEST CONTIGUOUS strictly increasing sequence.
# Print the sequence and its length.
# If multiple sequences have the same maximum length,
# print the FIRST such sequence.
#
# Do NOT use max(), sorted(), or dictionaries.

numbers = [3, 5, 7, 2, 4, 6, 8, 1, 3, 5]

current=[numbers[0]]
longest=[numbers[0]]

for i in range(1,len(numbers)):
    if numbers[i]>numbers[i-1]:
        current.append(numbers[i])
    else:
        current=[numbers[i]]

    if len(current)>len(longest):
        longest=current.copy()

print(longest)
```

```python
# DAY 33 - QUESTION 5
#
# Compress consecutive repeated numbers.
# Create two lists:
# values -> each consecutive value once
# freq   -> how many times it repeats consecutively
#
# Example: [2, 2, 2, 5, 5, 3, 3, 2]
# values = [2, 5, 3, 2]
# freq   = [3, 2, 2, 1]
#
# Do NOT use dictionaries, .count(), or Counter().

numbers = [4, 4, 4, 2, 2, 7, 7, 7, 1, 4, 4, 9]

values = []
freq = []

for i in numbers:
    if len(values) == 0 or i != values[-1]:
        values.append(i)
        freq.append(1)
    else:
        freq[-1] += 1

print(values)
print(freq)
```

```python
# DAY 33 - QUESTION 6
#
# For each distinct number in list a:
# 1. Count its frequency in a.
# 2. Count its frequency in b.
# 3. Print numbers that appear MORE often in a than in b.
# 4. Print each number only once.
# 5. Preserve first-appearance order from a.
#
# Do NOT use .count(), set(), Counter(), or dictionaries.

a = [3, 5, 3, 2, 5, 3, 7, 2, 8, 5]
b = [3, 5, 2, 5, 8, 8, 9]

l1=[]
freq_a=[]
l2=[]
freq_b=[]

for i in a:
    if i not in l1:
        l1.append(i)
        freq_a.append(1)
    else:
        index=l1.index(i)
        freq_a[index]+=1

print("List A :")
print(l1)
print(freq_a)

for i in b:
    if i not in l2:
        l2.append(i)
        freq_b.append(1)
    else:
        index=l2.index(i)
        freq_b[index]+=1

print("List B :")
print(l2)
print(freq_b)


result = []

for i in range(len(l1)):
    number = l1[i]
    frequency_a = freq_a[i]

    if number in l2:
        index_b = l2.index(number)
        frequency_b = freq_b[index_b]
    else:
        frequency_b = 0

    if frequency_a > frequency_b:
        result.append(number)

print("Numbers appearing more often in a:", result)
```

```python
# DAY 33 - QUESTION 9
# Given two lists a and b:
# 1. For each distinct number in a, calculate its frequency in a.
# 2. Calculate its frequency in b.
# 3. Print numbers whose frequency is the SAME in both lists.
# 4. Print each number only once.
# 5. Preserve first-appearance order from a.
# 6. Ignore numbers that appear only in b.
# 7. Do not use .count(), set(), Counter(), or dictionaries.

a = [2, 4, 2, 7, 4, 9]
b = [4, 2, 8, 4, 2, 5]
result = []

for i in range(len(a)):
    frequency_a = 0
    frequency_b = 0

    for j in range(len(a)):
        if a[i] == a[j]:
            frequency_a += 1

    for j in range(len(b)):
        if a[i] == b[j]:
            frequency_b += 1

    already_added = False
    for k in range(len(result)):
        if result[k] == a[i]:
            already_added = True
            break

    if frequency_a == frequency_b and not already_added:
        result.append(a[i])

print(result)
```
