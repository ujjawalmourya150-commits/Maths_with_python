```python
# 1. Find all duplicates — without repeating them

Numbers = [1,3,2,4,3,7,2,4,1,5,6]

def duplicate(n):
    
    seen = []

    for i in range(len(n)):
        for j in range(i+1,len(n)):
            if n[i] == n[j] and n[i] not in seen:
                seen.append(n[i])
    return seen
duplicate(Numbers)
```

```python
# 2. Find duplicate numbers and their counts

Numbers = [1,3,2,4,3,7,2,4,1,5,6]

def duplicate(n):
    seen = []

    for i in range(len(n)):
        count = 0

        for j in range(len(n)):
            if n[i] == n[j]:
                count += 1

        if count > 1 and n[i] not in seen:
            print(f"{n[i]} occurs {count} ")
            seen.append(n[i])

duplicate(Numbers)
```

```python
# 3. Find the number that appears the most

def duplicate(n):

    max_count = 0
    max_number = None

    for i in range(len(n)):
        count = 0

        for j in range(len(n)):

            if n[i] == n[j]:
                count += 1

        if count > max_count:

            max_count = count
            max_number = n[i]

    print(f"{max_number} appears {max_count} times")


duplicate([1,3,2,3,2,3])
```

```python
# 5. Find the first non-repeated number

def non_repeated(n):
    for i in range(len(n)):
        count = 0

        for j in range(len(n)):
            if n[i] == n[j]:
                count += 1

        if count == 1:
            return n[i]

print(non_repeated([1,2,1,3,4,3]))
```

```python
# 6 . Find numbers occurring exactly twice

def twice(n):

    seen = []

    for i in range(len(n)):
        count = 0
        for j in range(len(n)):
            if n[i] == n[j]:
                count += 1

        
        if count == 2 and n[i] not in seen:
            print(n[i])
            seen.append(n[i])

twice([1,2,2,3,3,1,4,5])
```

```python
# 9. Find the duplicate with the highest frequency


def highest_duplicate(n):
    highest_count = 0
    max_number = None
    
    for i in range(len(n)):
        count = 0
        for j in range(len(n)):
            if n[i] == n[j]:
                count+=1
        if count > highest_count:
            highest_count = count
            max_number = n[i]

    return f"The number with most frequency is {max_number}"

print(highest_duplicate([2, 5, 3, 5, 2, 5, 4, 3]))
```

```python
# 10. Find the last duplicate

def last_duplicate(n):
    for i in range(len(n) - 1, -1, -1):
        for j in range(i - 1, -1, -1):
            if n[i] == n[j]:
                return n[i]

print(last_duplicate([1, 2, 3, 2, 1, 4, 3]))
```
