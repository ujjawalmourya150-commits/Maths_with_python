```python
# 1 . find the second-highest frequency number.


def duplicate_latest_second(n):
    highest_freq = 0
    second_highest = 0
    max_number = None
    second_max = None

    for i in range(len(n)):
        count = 0
        for j in range(len(n)):
            if n[i] == n[j]:
                count+=1

        if count > highest_freq:
            second_highest = highest_freq
            highest_freq = count
            second_max = max_number
            max_number = n[i]
            
        elif count > second_highest and count != highest_freq:
            second_max = n[i]

    return second_max

print(duplicate_latest_second([2, 5, 3, 2, 7, 5, 8, 3, 5]))
```
