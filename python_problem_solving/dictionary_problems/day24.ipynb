```python
# 1 . Given a dictionary of students and marks, 
# create a new dictionary containing only students who scored 50 or more.

students = {
    "Rahul": 45,
    "Aman": 78,
    "Rohit": 52,
    "Karan": 39,
    "Mohit": 65
}
d = {}


# with use of .items()
for name,marks in students.items():
    if marks >= 50:
        d[name]=marks

print(d)


# without use of .items()
for name in students:
    if students[name] >= 50:
        d[name]=students[name]
print(d)
```

```python
# 2 . Given a dictionary of students and marks, 
# create a new dictionary where every student's marks are increased by 5.

students = {
    "Rahul": 45,
    "Aman": 70,
    "Rohit": 62,
    "Karan": 80
}

d = {}

d = {}

for name in students:
    d[name] = students[name] + 5

print(d)
```

```python
# 3 . Find the student with the lowest marks

students = {
    "Rahul": 72,
    "Aman": 45,
    "Rohit": 81,
    "Karan": 39,
    "Mohit": 65
}

lowest = None

for i in students:
    if lowest is None or  students[i] < lowest:
        lowest = students[i]
print(lowest) 
```

```python
# 4 . Group words by their first letter

words = [
    "apple", "ant", "banana",
    "ball", "cat", "car"
]

d = {}

d = {}

for i in words:
    first = i[0]

    if first in d:
        d[first].append(i)
    else:
        d[first] = [i]

print(d)
```

```python
# 5 . Given a product dictionary, find the product having the highest price.

products = {
    "Mouse": 500,
    "Keyboard": 1200,
    "Headphones": 1800,
    "Monitor": 7000,
    "USB": 400
}

highest = 0

for i in products:
    if products[i] > highest:
        highest = products[i]
print(highest)
```

```python
# 6 . Count the frequency of each word length
# Given a list of words, create a dictionary where:
#key → length of word
#value → number of words having that length

words = [
    "cat", "dog", "apple",
    "mango", "hi", "go", "bat"
]

d = {}

for i in words:
    l = len(i)
    if l in d:
        d[l]+=1
    else:
        d[l]=1
print(d)
```

```python
# 8 . Create a dictionary containing the total quantity sold for each product.

sales = [
    ("Laptop", 2),
    ("Phone", 3),
    ("Laptop", 1),
    ("Tablet", 4),
    ("Phone", 2),
    ("Laptop", 2)
]

d = {}

for product, quantity in sales:
    if product in d:
        d[product] += quantity
    else:
        d[product] = quantity

print(d)
```

```python
# 10 . Calculate the average marks of each student, then find the student with the highest average.

marks = [
    ("Rahul", 70),
    ("Aman", 80),
    ("Rahul", 90),
    ("Aman", 60),
    ("Rohit", 75),
    ("Rohit", 85)
]

d = {}
count = {}

for n, m in marks:
    if n in d:
        d[n] += m
        count[n] += 1
    else:
        d[n] = m
        count[n] = 1

highest = 0
student = ""

for n in d:
    average = d[n] / count[n]

    if average > highest:
        highest = average
        student = n

print(student)
print(highest)
```
