```python
# 1. Calculate total marks for each student.
# Find ALL students with the SECOND-HIGHEST DISTINCT total.
# Students may appear multiple times.


names = ["Aman", "Rahul", "Rohit", "Aman",
         "Rahul", "Rohit", "Karan"]
marks = [80, 70, 90, 40, 50, 20, 100]

d={}

highest=0
second=0
h_num=None
s_num=None


for i in range(len(names)):
    if names[i] in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]
print(d)
for i in d:
    if d[i]>highest:
        second=highest
        s_num=h_num
        highest=d[i]
        h_num=i
    elif d[i]>second and d[i]!=highest:
        second=d[i]
        s_num=i
print(s_num)
```

```python
# 2. Count the frequency of each product.
# Create a new dictionary:
# key   -> frequency
# value -> list of products with that frequency
# Keep the original order of first appearance.

products = ["Mouse", "Keyboard", "Mouse",
            "Laptop", "Keyboard", "Phone",
            "Mouse", "Phone", "Tablet"]
d={}

result = {}

for i in products:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i] in result:
        result[d[i]].append(i)
    else:
        result[d[i]]=[i]
print(result)
```

```python
# 3. Find the SECOND-HIGHEST DISTINCT frequency.
# Return all numbers having that frequency.
# Do NOT use count(), max(), min(), sorted(), set(), or Counter.

numbers = [2, 2, 2, 3, 3, 4, 4, 5, 5, 6]

d ={}
highest=0
second=0
result = []


for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i]>highest:
        second=highest
        highest=d[i]
        
    elif d[i]>second and d[i]!=highest:
        second=d[i]
        
for i in d:
    if d[i]==second:
        result.append(i)
print(f"The SECOND-HIGHEST DISTINCT frequency are {result}")
```

```python
# 4. Calculate each student's total marks.
# Then classify students into groups:
# key   -> "High" if total >= 150
#         "Medium" if total >= 100
#         "Low" otherwise
# value -> list of student names
# Each student may appear multiple times.

names = ["Aman", "Rahul", "Rohit", "Aman",
         "Rahul", "Karan", "Rohit"]
marks = [80, 60, 70, 90, 50, 40, 50]

d={}

result={"High":[],
        "Medium":[],
        "Low":[]}

for i in range(len(names)):
    if names[i] in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]

for i in d:
    if d[i] >= 150:
        result["High"].append(i)
    elif d[i] >= 100:
        result["Medium"].append(i)
    else:
        result["Low"].append(i)
print(result)
```

```python
# 5. Calculate total quantity sold for each product.
# Calculate the average of the product totals.
# Print all products whose total is ABOVE the average.
# Products may appear multiple times.

products = ["Mouse", "Keyboard", "Mouse", "Laptop",
            "Keyboard", "Laptop", "Phone"]
quantities = [10, 5, 5, 20, 5, 10, 5]

d = {}
total=0
result = []

for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=quantities[i]
    else:
        d[products[i]]=quantities[i]
print(d)
for i in d:
    l=len(d)
    total+=d[i]
avg=total/l
print(f"The average of products total is {avg}")

for i in d:
    if d[i]>avg:
        result.append(i)
print(result)
```

```python
# 7. For each student, find their HIGHEST individual mark.
# Create a dictionary:
# key   -> highest mark
# value -> list of students with that highest mark
# Do NOT calculate total marks.

names = ["Aman", "Rahul", "Aman", "Rohit",
         "Rahul", "Rohit", "Karan"]
marks = [80, 70, 95, 85, 90, 85, 95]
d={}
result={}

names = ["Aman", "Rahul", "Aman", "Rohit",
         "Rahul", "Rohit", "Karan"]
marks = [80, 70, 95, 85, 90, 85, 95]

d = {}
result = {}

for i in range(len(names)):
    if names[i] not in d:
        d[names[i]] = marks[i]
    elif marks[i] > d[names[i]]:
        d[names[i]] = marks[i]

print(d)

for i in d:
    if d[i] in result:
        result[d[i]].append(i)
    else:
        result[d[i]] = [i]

print(result)
```

```python
# 8. Calculate total quantity sold for each product.
# Find ALL products tied for the LOWEST total.
# Do NOT use min(), max(), sorted(), set(), or Counter.

products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Keyboard", "Mouse"]
quantities = [5, 10, 3, 5, 2, 2, 3]

d={}
lowest=None
answer=[]

for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=quantities[i]
    else:
        d[products[i]]=quantities[i]
print(d)
for i in d:
    if lowest is None or d[i]<lowest:
        lowest=d[i]
for i in d:
    if lowest==d[i]:
        answer=[i]
print(answer)
```

```python
# 9. Group words by their length.
# Then find ALL word lengths having the highest
# number of words.
# Do NOT use max(), min(), sorted(), set(), or Counter.

words = ["cat", "dog", "sun", "apple",
         "mango", "bat", "grape", "kiwi", "lion"]
d={}

for i in words:
    l=len(i)
    if l in d:
        d[l]+=[i]
    else:
        d[l]=[i]
print(d)

highest = 0
answer = []

for i in d:
    if len(d[i]) > highest:
        highest = len(d[i])

for i in d:
    if len(d[i]) == highest:
        answer.append(i)

print(answer)
```

```python
# 10. Calculate total sales amount for each product.
# Then group products by their total sales.
# Finally, find ALL products with the SECOND-HIGHEST
# DISTINCT sales total.


products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Keyboard", "Mouse", "Keyboard"]

amounts = [500, 800, 200, 300, 400, 600, 100, 100]

d={}
highest=0
second=0

for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=amounts[i]
    else:
        d[products[i]]=amounts[i]
print(d)

for i in d:
    if d[i]>highest:
        second=highest
        highest=d[i]
    elif d[i]>second and d[i]!=second:
        second=d[i]
        
answer = []

for i in d:
    if d[i] == second:
        answer.append(i)

print(answer)
```
