```python
# DAY 29 - QUESTION 1
#
# 1. Calculate total sales for each product.
# 2. Count how many times each product appeared.
# 3. Find the HIGHEST distinct sales total.
# 4. Print products having that total ONLY if
#    they appeared at least twice.
#
# Do NOT use max(), min(), sorted(), set(),
# or Counter().

products = ["Phone", "Laptop", "Phone", "Mouse",
            "Laptop", "Phone", "Mouse", "Tablet",
            "Tablet", "Keyboard"]

amounts = [300, 500, 200, 150, 400, 100,
           250, 350, 150, 600]

d={}
highest=0

for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=amounts[i]
    else:
        d[products[i]]=amounts[i]
print(d)

count={}

for i in range(len(products)):
    if products[i] in count:
        count[products[i]]+=1
    else:
        count[products[i]]=1
print(count)

for i in d:
    if d[i]>highest:
        highest=d[i]

found=False

for i in d:
    if count[i]>=2 and d[i]==highest:
        print(f"The product that appears twice and is highest total sales is {i} ")
        found=True

if not found:
    print("No product qualifies.")
```

```python
# DAY 29 - QUESTION 2
#
# 1. Count the frequency of every number.
# 2. Find the SECOND-HIGHEST DISTINCT frequency.
# 3. Print ALL numbers having that frequency.
# 4. Print them in their original first-appearance
#    order.
#
# Do NOT use max(), min(), sorted(), set(),
# Counter(), or count().

numbers = [4, 2, 4, 3, 2, 5, 4, 3, 6,
           2, 5, 3, 7, 6, 8, 6, 8]

d = {}
highest=0
second=0


for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)

for i in d:
    if d[i]>highest:
        second=highest
        highest=d[i]
    elif d[i]>second and d[i]!=highest:
        second=d[i]

result = []

for i in numbers:
    if d[i] == second and i not in result:
        result.append(i)

print(result)
```

```python
# DAY 29 - QUESTION 3
#
# 1. Group students and their marks by subject.
# 2. For EACH subject, find the SECOND-HIGHEST
#    DISTINCT mark.
# 3. Print ALL students who obtained that mark.
#
# Do NOT use max(), min(), sorted(), or set().


students = ["Aman", "Riya", "Rohit", "Aman",
            "Karan", "Riya", "Karan", "Rohit",
            "Aman", "Karan"]

subjects = ["Maths", "Maths", "Maths", "Science",
            "Science", "Science", "Maths", "Science",
            "Maths", "Science"]

marks = [90, 80, 80, 70, 95, 85, 90, 85, 70, 75]

d = {}

for i in range(len(subjects)):
    if subjects[i] in d:
        d[subjects[i]].append([students[i], marks[i]])
    else:
        d[subjects[i]] = [[students[i], marks[i]]]

print("Grouped data:", d)

for subject in d:
    highest = 0
    second = 0

    for student, mark in d[subject]:
        if mark > highest:
            second = highest
            highest = mark
        elif mark > second and mark != highest:
            second = mark

    print(f"\nSubject: {subject}")
    print(f"Second-highest distinct mark: {second}")

    for student, mark in d[subject]:
        if mark == second:
            print(student, mark)

```

```python
# DAY 29 - QUESTION 4
#
# 1. Calculate total sales for each product.
# 2. Group products into these categories:
#       High   -> total sales >= 800
#       Medium -> total sales >= 400 and < 800
#       Low    -> total sales < 400
# 3. Store product names under their categories.
# 4. Print the final category dictionary.
#
# Do NOT use max(), min(), sorted(), or set().

products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Mouse", "Keyboard", "Phone",
            "Keyboard", "Tablet"]

amounts = [300, 500, 200, 400, 600, 150, 250,
           200, 300, 350]
d={}
result={"high":[],
       "medium":[],
       "low":[]}


for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=amounts[i]
    else:
        d[products[i]]=amounts[i]
print(d)

for i in d:
    if d[i]>=800:
        result["high"].append(i)
    elif d[i]>=400 and d[i]<800:
        result["medium"].append(i)
    else:
        result["low"].append(i)
print(result)
```

```python
# DAY 29 - QUESTION 5
#
# 1. Calculate total sales for each product.
# 2. Calculate the average of the DISTINCT
#    products' total sales.
# 3. Print ALL products whose total sales are
#    strictly greater than the average.
#
# Do NOT use sum(), max(), sorted(), set(),
# or Counter().


products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Keyboard", "Mouse", "Tablet",
            "Tablet", "Phone"]

amounts = [400, 600, 200, 300, 200, 350, 100,
           500, 150, 100]

d = {}

for i in range(len(products)):
    if products[i] in d:
        d[products[i]] += amounts[i]
    else:
        d[products[i]] = amounts[i]

print("Total sales:", d)

total = 0

for i in d:
    total += d[i]

count = 0

for i in d:
    count += 1

average = total / count

print("Average sales:", average)

for i in d:
    if d[i] > average:
        print(i, d[i])
    
```

```python
# DAY 29 - QUESTION 8
#
# 1. Calculate total marks for each student.
# 2. Find the SECOND-HIGHEST DISTINCT student total.
# 3. Count how many subjects each student has
#    a mark recorded for.
# 4. Print students with the second-highest total
#    ONLY if they have marks recorded at least
#    three times.
#
# Do NOT use max(), min(), sorted(), set(),
# or Counter().

students = ["Aman", "Riya", "Rohit", "Aman",
            "Karan", "Riya", "Rohit", "Aman",
            "Karan", "Riya", "Rohit", "Karan"]

marks = [80, 90, 70, 85, 75, 80, 90, 70,
         95, 85, 80, 85]

d ={}
count={}
highest=0
second=0


for i in range(len(students)):
    if students[i] in d:
        d[students[i]]+=marks[i]
        count[students[i]]+=1
    else:
        d[students[i]]=marks[i]
        count[students[i]]=1

print(d)
print(f"\nSubjects each student has a mark recorded for {count}")

for i in d:
    if d[i]>highest:
        second=highest
        highest=d[i]
    elif d[i]>second and d[i]!=highest:
        second=d[i]
print(f"\nThe SECOND-HIGHEST DISTINCT student total is {second}")

result=[]

for i in d:
    if d[i]==second and count[i] >= 3:
        result.append(i)

if result:
    print("\nQualifying students:", result)
else:
    print("\nNo students qualify.")
```

```python
# DAY 29 - QUESTION 9
#
# 1. Calculate total sales for each product.
# 2. Group products by category.
# 3. For EACH category, find the SECOND-HIGHEST
#    DISTINCT product sales total.
# 4. Print ALL products having that total in
#    their corresponding category.
#
# Do NOT use max(), min(), sorted(), set(),
# or Counter().


products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Mouse", "Tablet", "Tablet",
            "Keyboard", "Keyboard", "Phone", "Mouse"]

categories = ["Electronics", "Electronics", "Accessories",
              "Electronics", "Electronics", "Accessories",
              "Electronics", "Electronics", "Accessories",
              "Accessories", "Electronics", "Accessories"]

amounts = [300, 500, 200, 400, 600, 100,
           250, 350, 150, 250, 100, 300]

d = {}
d1 = {}


for i in range(len(products)):
    product = products[i]
    category = categories[i]
    amount = amounts[i]

    if product in d:
        d[product] += amount
    else:
        d[product] = amount

    if category in d1:
        if product not in d1[category]:
            d1[category].append(product)
    else:
        d1[category] = [product]

print("Total sales:", d)
print("Products by category:", d1)

for category in d1:
    highest = 0
    second = 0

    for product in d1[category]:
        total = d[product]

        if total > highest:
            second = highest
            highest = total
        elif total > second and total != highest:
            second = total

    print(f"\nCategory: {category}")
    print("Second-highest distinct sales total:", second)

    result = []

    for product in d1[category]:
        if d[product] == second:
            result.append(product)

    if result:
        print("Matching products:", result)
    else:
        print("No matching products found.")
```

```python
# DAY 29 - QUESTION 10
#
# 1. Calculate total sales for each product.
# 2. Count each product's appearances.
# 3. Group products by their sales category:
#       High   -> total >= 700
#       Medium -> total >= 400 and < 700
#       Low    -> total < 400
# 4. For EACH category, find the SECOND-HIGHEST
#    DISTINCT total sales.
# 5. Print products having that total ONLY if
#    they appeared at least twice.
# 6. If no product qualifies in a category,
#    print an appropriate message for that category.
#
# Do NOT use max(), min(), sorted(), set(),
# Counter(), or sum().
products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Keyboard", "Mouse", "Phone",
            "Tablet", "Keyboard", "Tablet", "Mouse",
            "Laptop", "Phone"]

amounts = [300, 500, 200, 400, 200, 150, 100,
           100, 350, 250, 150, 100, 100, 200]

d = {}
count = {}
groups = {
    "high": [],
    "medium": [],
    "low": []
}

for i in range(len(products)):
    if products[i] in d:
        d[products[i]] += amounts[i]
        count[products[i]] += 1
    else:
        d[products[i]] = amounts[i]
        count[products[i]] = 1

print("Total sales:", d)
print("Product Appearances:", count)


for product in d:
    if d[product] >= 700:
        groups["high"].append(product)
    elif d[product] >= 400:
        groups["medium"].append(product)
    else:
        groups["low"].append(product)

print("\nProducts by category:", groups)


for category in groups:
    highest = 0
    second = 0

    for product in groups[category]:
        total = d[product]

        if total > highest:
            second = highest
            highest = total
        elif total > second and total != highest:
            second = total

    print(f"\nCategory: {category}")

    if second == 0:
        print("No second-highest distinct total exists.")
        continue

    print("Second-highest distinct total:", second)

    result = []

    for product in groups[category]:
        if d[product] == second and count[product] >= 2:
            result.append(product)

    if result:
        print("Qualifying products:", result)
    else:
        print("No product qualifies in this category.")
```
