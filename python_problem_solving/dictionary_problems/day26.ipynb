```python
# 1. Given a list of numbers,
# create a dictionary where:
#
# key   -> "positive", "negative", or "zero"
# value -> list of numbers belonging to that group

numbers = [4, -2, 0, 7, -5, 0, 9, -1]
d = {"postivie":[],
     "negative":[],
     "zero":[]}


for i in numbers:
    if i>0:
        d["postivie"].append(i)
    elif i == 0:
        d["zero"].append(i)
    else:
        d["negative"].append(i)
print(d)
```

```python
# 2. Given a list of words,
# group words having the same length.
#
# key   -> length of the word
# value -> list of words with that length

words = ["cat", "dog", "apple", "bat", "mango", "sun"]
d = {}
# Expected:
# {
#     3: ["cat", "dog", "bat", "sun"],
#     5: ["apple", "mango"]
# }
for i in words:
    l = len(i)
    if l in d:
        d[l].append(i)
    else:
        d[l]=[i]
print(d)
```

```python
# 3. Given a list of product names,
# count how many times each product appears.
#
# key   -> product name
# value -> number of appearances

products = [
    "Mouse", "Keyboard", "Mouse",
    "Monitor", "Keyboard", "Mouse"
]

d = {}

for i in products:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
# 4. Given a list of student names,
# create a dictionary containing the number
# of times each student appears.
#
# Then print only students who appear
# more than once.

students = [
    "Rahul", "Aman", "Rohit",
    "Rahul", "Karan", "Aman", "Rahul"
]

d = {}

for i in students:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)

for i in d:
    if d[i]>1:
        print(i)
```

```python
# 5. Given two lists containing student names and marks,
# calculate the TOTAL marks of each student.
# Students may appear multiple times.

names = ["Aman", "Rahul", "Aman", "Rohit", "Rahul"]
marks = [60, 75, 80, 65, 85]

d={}


for i in range(len(names)):
    if names[i] in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]
print(d)
```

```python
# 6. Given two lists,
# calculate the total quantity sold for each product.
#
# products[i] matches quantities[i].
# A product may appear multiple times.

products = ["Mouse", "Keyboard", "Mouse", "Monitor", "Keyboard"]
quantities = [3, 2, 5, 1, 4]

d={}

for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=quantities[i]
    else:
        d[products[i]]=quantities[i]
print(d)
```

```python
# 7. Given two lists of student names and marks,
# first calculate the total marks of every student.
# Then find the student with the HIGHEST total.
#
# Students may appear multiple times.

names = ["Rahul", "Aman", "Rohit", "Rahul", "Aman", "Rohit"]
marks = [60, 70, 85, 55, 80, 40]

d = {}
highest=0
answer=None


for i in range(len(names)):
    if names[i] in d:
        d[names[i]] += marks[i]
    else:
        d[names[i]] = marks[i]
print(d)
for i in d:
    if d[i]>highest:
        highest=d[i] 
        answer=i
print(f"{answer} got {highest} which is highest marks. ")
```

```python
# 8. Given a list of numbers,
# find the FIRST number in the original list
# that appears exactly 2 times.
#
# First create a dictionary of frequencies.
# Then find the answer in the original list order.
#
# Do NOT use count(), Counter, or set().

numbers = [4, 2, 5, 2, 4, 7, 5, 4, 8]
d ={}


for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d) 
for i in d:
    if d[i]==2:
        print(f"{i} is the number who appears exactly twice.")
        break
```

```python
# 9. Given a list of product names,
# find ALL products having the LOWEST frequency.
#
# If multiple products have the same lowest frequency,
# return all of them in a list.


products = [
    "Mouse", "Keyboard", "Mouse",
    "Monitor", "Keyboard", "Laptop"
]

lowest = None
answer = []
d={}

for i in products:
    if i in d:
        d[i]+=1
    else:
        d[i]=1


for i in d:
    if lowest is None or d[i]<lowest:
        lowest=d[i]
        answer=[i]
    elif d[i] == lowest:
        answer.append(i)
print(f"The lowest frequency numbers are {answer}")


```

```python
# 10. Given two lists containing student names and marks:
#
# Step 1: Calculate the TOTAL marks of each student.
# Step 2: Create a NEW dictionary where:
#
# key   -> total marks
# value -> list of students having that total
#
# Students may appear multiple times.

names = ["Rahul", "Aman", "Rohit", "Rahul", "Aman", "Karan"]
marks = [60, 70, 80, 40, 30, 100]

d = {}
result = {}

for i in range(len(names)):
    if names[i]in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]

for i in d:
    if d[i] in result:
        result[d[i]].append(i)
    else:
        result[d[i]] = [i]
print(result)
```
