```python
# 1 . Given a dictionary of products and prices,
# create a new dictionary containing only products
# whose price is greater than 500.

products = {
    "Mouse": 400,
    "Keyboard": 800,
    "Headphones": 1200,
    "USB": 300,
    "Monitor": 7000
}

new_dict = {}

for product,price in products.items():
    if price > 500:
        new_dict[product]=price
print(new_dict)
```

```python
# 2 . Given a list of numbers, create a dictionary
# where the key is the number and the value is
# how many times it appears.

numbers = [2, 3, 2, 5, 3, 2, 7, 5, 5]

new_dict = {}

for i in numbers:
    if i in new_dict:
        new_dict[i]+=1
    else:
        new_dict[i]=1
print(new_dict)
```

```python
# 3 . Given a list of numbers, create a dictionary
# containing only numbers that appear more than 2 times.

numbers = [2, 3, 2, 5, 3, 2, 7, 5, 5, 5, 8]

new_dict = {}
answer = {}

for i in numbers:
    if i in new_dict:
        new_dict[i] += 1
    else:
        new_dict[i] = 1

for i in new_dict:
    if new_dict[i] > 2:
        answer[i] = new_dict[i]

print(answer)
```

```python
# 4. Find the number with the lowest frequency.
# If two numbers have the same frequency,
# return the one that appears first in the original list.

numbers = [4, 2, 4, 7, 2, 9, 7, 2]

d = {}

lowest_freq = 999
answer = None

for i in numbers:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

for i in d:
    if d[i] < lowest_freq:
        lowest_freq = d[i]
        answer = i

print(answer)
```

```python
# 5. Find all numbers that have the highest frequency.

numbers = [2, 2, 3, 3, 4, 4, 5]

new_dict = {}

highest = 0
answer = []

for i in numbers:
    if i in new_dict:
        new_dict[i]+=1
    else:
        new_dict[i]=1

for i in new_dict:
    if new_dict[i] > highest:
        highest=new_dict[i]
        answer=[i]
  
    elif new_dict[i] == highest:
        answer.append(i)

print(answer)
```

```python
# 6 .Create two dictionaries:

# unique   -> numbers appearing exactly once
# duplicate -> numbers appearing more than once
# The value should be the frequency.

numbers = [2, 3, 2, 5, 3, 7, 8, 8, 9]

d={}
unique ={}
duplicate ={}

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i]==1:
        unique[i]=d[i]
    elif d[i]>1:
        duplicate[i]=d[i]
print(unique)
print(duplicate)
```

```python
# 7 . A student can appear multiple times.
# Calculate the total marks for each student,
# then find the student with the highest total.

students = [
    ("Rahul", 50),
    ("Aman", 80),
    ("Rahul", 70),
    ("Rohit", 60),
    ("Aman", 90),
    ("Rohit", 50)
]

d = {}

highest = 0
answer = None

for name,marks in students:
    if name in  d:
        d[name]+=marks
    else:
        d[name]=marks

for i in d:
    if d[i]>highest:
        highest=d[i]
        answer=i 
print(answer)
```

```python
# 8 . Given a dictionary of students and marks,
# create a new dictionary where:
# key   -> marks
# value -> list of students who got those marks

students = {
    "Rahul": 80,
    "Aman": 90,
    "Rohit": 80,
    "Karan": 70,
    "Mohit": 90
}

d = {}

for name, marks in students.items():
    if marks in d:
        d[marks].append(name)
    else:
        d[marks] = [name]

print(d)
```

```python
# 9 . A product can appear multiple times.
# Calculate total sales for each product,
# then find the product with the highest total sales.


sales = [
    ("Laptop", 50000),
    ("Phone", 20000),
    ("Laptop", 30000),
    ("Tablet", 15000),
    ("Phone", 25000),
    ("Tablet", 10000)
]

d = {}
highest=0
answer = None

for item,price in sales:
    if item  in d:
        d[item]+=price
    else:
        d[item]=price
print(d)

for i in d:
    if d[i]>highest:
        highest=d[i]
        answer=i
print(answer)
```
