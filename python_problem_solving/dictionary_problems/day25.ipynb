```python
# 1. Given a dictionary of students and marks,
# create a new dictionary containing only students
# who scored 70 or more.

students = {
     "Rahul": 65,
     "Aman": 82,
     "Rohit": 74,
     "Karan": 55,
     "Mohit": 91
 }

d = {}
for name in students:
    if students[name]>70:
        d[name]=students[name]

print(d)
```

```python
# 2. Given a dictionary of students and marks,
# increase every student's marks by 10
# and create a new dictionary.

students = {
     "Rahul": 60,
     "Aman": 75,
     "Rohit": 80,
     "Karan": 65
}

d = {}

for name in students:
    students[name]+=10
    d[name]=students[name]
print(d) 
```

```python
# # 3. Given a dictionary of students and marks,
# find the student who scored the LOWEST marks.
#
# Do NOT use min().

students = {
     "Rahul": 65,
     "Aman": 82,
     "Rohit": 54,
     "Karan": 71,
     "Mohit": 60
 }

d ={}
lowest = None
answer = None

for name in students:
    if lowest is None or students[name]<lowest:
        lowest = students[name]
        answer = name
print(f"{name} scored {lowest} ,which is the lowest marks.")
```

```python
# 4. Given a list of words,
# create a dictionary where:
#
# key   -> length of the word
# value -> number of words having that length

words = ["cat", "dog", "apple", "bat", "mango", "sun"]

d = {}
for i in words:
    l = len(i)
    if l in d:
        d[l]+=1
    else:
        d[l]=1
print(d)
```

```python
# 5. Given a list of numbers,
# create a dictionary where:
#
# key   -> "even" or "odd"
# value -> list of numbers belonging to that group

numbers = [2, 5, 8, 11, 14, 17, 20]

d = {
    "even":[],
    "odd":[]
}

for i in numbers:
    if i % 2==0:
        d["even"].append(i)
    else:
        d["odd"].append(i)
print(d)
```

```python
# 6. Given a list of words,
# group the words by their FIRST LETTER.
#
# key   -> first letter of the word
# value -> list of words starting with that letter

words = ["apple", "ant", "banana", "ball", "cat", "car"]

d = {}

for i in words:
    f = i[0]
    if f in d:
        d[f].append(i)
    else:
        d[f]=[i]
print(d)
```

```python
# 7. Given a dictionary of products and prices,
# find the product having the LOWEST price.
#
# Do NOT use min().

products = {
     "Mouse": 500,
     "Keyboard": 1200,
     "Headphones": 1800,
     "USB": 400,
     "Webcam": 900
}

d = {}
lowest = None
answer = None

for item in products:
    if lowest is None or products[item]<lowest:
        lowest=products[item]
        answer = item
print(f"{answer} price is {lowest} and is the lowest priced product. ")
```

```python
# 8. Given two lists containing student names and marks,
# some students appear more than once.
#
# Create a dictionary containing the TOTAL marks
# scored by each student.


names = ["Rahul", "Aman", "Rahul", "Rohit", "Aman"]
marks = [70, 80, 65, 75, 90]


names = ["Rahul", "Aman", "Rahul", "Rohit", "Aman"]
marks = [70, 80, 65, 75, 90]

d = {}

for i in range(len(names)):
    if names[i] in d:
        d[names[i]] += marks[i]
    else:
        d[names[i]] = marks[i]

print(d)
```

```python
# 9. Given a list of numbers,
# find the FIRST number in the original list
# that appears MORE THAN 2 times.

numbers = [5, 2, 3, 2, 5, 7, 3, 5, 2, 8]
d = {}
# Frequencies:
# 5 -> 3
# 2 -> 3
# 3 -> 2
# 7 -> 1
# 8 -> 1

# Expected:
# 5

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i]>2:
        print(i)
        break
```

```python
# 10. Given two lists containing student names and marks,
# students can appear multiple times.
# First calculate the TOTAL marks of each student.
# Then find the student having the HIGHEST total marks.

names = ["Rahul", "Aman", "Rohit", "Rahul", "Aman", "Rohit"]
marks = [70, 80, 90, 60, 75, 50]
d = {}
highest = 0
answer = None

for i in range(len(names)):
    if names[i] in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]

for i in d:
    if d[i]>highest:
        highest=d[i]
        answer=i
print(f"{answer} has the highest marks.")
```
