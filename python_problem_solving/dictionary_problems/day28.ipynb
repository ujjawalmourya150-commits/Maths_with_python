```python
# 1. Calculate the total marks of each student.
# Then find ALL students who have the
# SECOND-HIGHEST DISTINCT total marks.
#
# If multiple students have the same second-highest
# total, print all of them.

names = ["Aman", "Riya", "Aman", "Rohit",
         "Riya", "Karan", "Rohit", "Karan"]
marks = [70, 80, 90, 60, 85, 75, 80, 70]

d={}
highest=0
second=0
result=[]

for i in range(len(names)):
    if names[i] in d:
        d[names[i]]+=marks[i]
    else:
        d[names[i]]=marks[i]
print(d)

for i in d:
    if d[i]>highest:
        second=highest
        highest=d[i]
    elif d[i]>second and d[i]!=highest:
        second=d[i]

for i in d:
    if d[i] not in result:
        if d[i] == second:
            result.append(i)
print(result)
```

```python
# 4. Calculate the average marks of each student.
# Create a dictionary where:
# key   -> average marks
# value -> list of students with that average
#
# Each student should appear only once.
# If two students have the same average, group them together.

names = ["Aman", "Riya", "Aman", "Riya",
         "Karan", "Karan", "Rohit", "Rohit"]

marks = [80, 70, 90, 100, 60, 80, 75, 85]

d = {}
count = {}

# Step 1: Calculate total marks and count marks per student
for i in range(len(names)):
    if names[i] in d:
        d[names[i]] += marks[i]
        count[names[i]] += 1
    else:
        d[names[i]] = marks[i]
        count[names[i]] = 1

print("Total marks:", d)

# Step 2: Calculate average and group students
result = {}

for i in d:
    avg = d[i] / count[i]

    if avg in result:
        result[avg].append(i)
    else:
        result[avg] = [i]


    result[avg] = [i]
print("Average marks grouped by student:", result)
```

```python
# 6. Calculate the total sales for each product.
# Group products into these categories:
#
# "Low"    -> total below 400
# "Medium" -> total from 400 to 799
# "High"   -> total 800 or above
#
# Each product should appear only once.
# Do NOT use Counter() or set().

products = ["Phone", "Laptop", "Mouse", "Phone",
            "Laptop", "Mouse", "Tablet", "Tablet"]

amounts = [500, 700, 200, 300, 400, 100, 250, 350]

d={}

result={"Low": [],
    "Medium": [],
    "High": []}


for i in range(len(products)):
    if products[i] in d:
        d[products[i]]+=amounts[i]
    else:
        d[products[i]]=amounts[i]


for i in d:
    if d[i]>=800:
        result["High"].append(i)
    elif d[i]>=400 and d[i]<800:
        result["Medium"].append(i)
    else:
        result["Low"].append(i)
print(result)
```

```python
# 7. For each student, find their highest and
# lowest individual marks.
#
# Calculate:
# difference = highest mark - lowest mark
#
# Find ALL students with the smallest difference.
# Do NOT calculate total or average marks.


names = ["Aman", "Riya", "Aman", "Riya",
         "Karan", "Karan", "Rohit", "Rohit"]

marks = [80, 75, 85, 95, 60, 70, 88, 90]

d = {}
differences = {}


for i in range(len(names)):
    if names[i] in d:
        d[names[i]].append(marks[i])
    else:
        d[names[i]] = [marks[i]]


for student in d:
    highest = d[student][0]
    lowest = d[student][0]

    for mark in d[student]:
        if mark > highest:
            highest = mark

        if mark < lowest:
            lowest = mark

    difference = highest - lowest
    differences[student] = difference

    print(student, "Highest:", highest,
          "Lowest:", lowest,
          "Difference:", difference)


smallest = None

for student in differences:
    if smallest is None or differences[student] < smallest:
        smallest = differences[student]


print("\nStudents with the smallest difference:")

for student in differences:
    if differences[student] == smallest:
        print(student, smallest)
```

```python
# 8. Count the frequency of each number.
# Calculate the average frequency across all
# DISTINCT numbers.
#
# Print ALL numbers whose frequency is strictly
# greater than the average frequency.
#
# Do NOT use Counter(), max(), or sorted().

numbers = [2, 2, 2, 3, 3, 4, 4, 5, 6, 6, 6, 6]

d = {}

for i in numbers:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

print("Frequency:", d)


total = 0

for i in d:
    total += d[i]


average = total / len(d)

print("Average frequency:", average)


print("Numbers above average frequency:")

for i in d:
    if d[i] > average:
        print(i)
```

```python
# 9. Group students by subject.
# For EACH subject, find the SECOND-HIGHEST
# DISTINCT mark.
#
# Print ALL students who achieved that mark
# in the corresponding subject.


students = ["Aman", "Riya", "Rohit", "Aman",
            "Karan", "Riya", "Karan", "Rohit"]

subjects = ["Maths", "Maths", "Maths", "Science",
            "Science", "Science", "Maths", "Science"]

d={}

marks = [90, 80, 80, 70, 95, 85, 90, 85]

# Expected steps:
# 1. Group students and marks by subject.
# 2. Find the second-highest distinct mark in each subject.
# 3. Print all students with that mark.

for i in range(len(students)):
    if subjects[i] in d:
        d[subjects[i]].append((students[i],marks[i]))
    else:
        d[subjects[i]] = [(students[i], marks[i])]

for i in d:
    highest = 0
    second = 0

    for student, mark in d[i]:
        if mark > highest:
            second = highest
            highest = mark
        elif mark != highest and mark > second:
            second = mark

    
    print("\nSubject:", i)
    print("Second-highest distinct mark:", second)
    print("Students:")

    for student, mark in d[i]:
        if mark == second:
            print(student, mark)
```
