```python
# 1. Count the frequency of every element in a list.

numbers = [4, 2, 4, 5, 2, 4, 7, 5, 2, 8]

def frequency(n):

    freq = {}

    for i in n:

        if i in freq:
            freq[i]=+1

        else:
            freq[i]=1

    return freq
    
frequency(numbers)
```

```python
# 2 . Find the first element whose frequency is exactly 1.

numbers = [4, 7, 2, 7, 4, 9, 2, 5]

def frequency_1(n):

    freq = {}

    for i in n:

        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1

    for i in n:
        if freq[i] == 1:
            return i

frequency_1(numbers)
```

```python
# 3 .  Find the element with the highest frequency.
# If two elements have the same frequency, return the one
# that appears first in the original list.

numbers = [4, 2, 4, 7, 2, 4, 7, 2]

def frequency(n):

    highest = 0
    answer = None
    freq = {}


    for i in n:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1

    for i in n:
        if freq[i] > highest:
            highest = freq[i]
            answer = i
        
    return answer

frequency(numbers)
```

```python
# 4. Group numbers according to their frequency.

numbers = [1, 1, 2, 2, 2, 3]

def group_frequency(n):
    freq = {}
    groups = {}

    for i in n:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1

    for i in freq:
        f = freq[i]

        if f in groups: 
            groups[f].append(i)
        else:
            groups[f] = [i]

    return groups

print(group_frequency(numbers))
```

```python
# 5 . Given a list of words, create a dictionary where
# each word is a key and its length is the value.

words = ["Python", "AI", "Machine", "Learning", "ML"]

w = {}

for i in words:
    l = len(i)
    w[i]=l
    
print(w)
```

```python
# 6 .Find the word that occurs most frequently.
# If there is a tie, return the word that appears first.

words = [ "AI","Python", "ML", "AI", "Python","AI", "ML"]

freq = {}

for i in words:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

most_freq = 0
answer = None

for i in words:
    if freq[i] > most_freq:
        most_freq = freq[i]
        answer = i

print(answer)
```

```python
# 7. Given student names and marks, create a dictionary
# and find the student who scored the highest marks.

names = ["Rahul", "Aman", "Rohit", "Ujjawal"]
marks = [72, 89, 81, 95]

new_dict = {}

for i in range(len(names)):
    new_dict[names[i]] = marks[i]

highest = 0
student = None

for i in new_dict:
    if new_dict[i] > highest:
        highest = new_dict[i]
        student = i

print(new_dict)
print(student)
print(highest)
```

```python
# 8. Given student names and marks, create a dictionary
# and find the student who scored the highest marks using zip().

names = ["Rahul", "Aman", "Rohit", "Ujjawal"]
marks = [72, 89, 81, 95]

new_dict = {}

for name, mark in zip(names, marks):
    new_dict[name] = mark

highest = 0
student = None

for name in new_dict:
    if new_dict[name] > highest:
        highest = new_dict[name]
        student = name

print(new_dict)
print(student)
print(highest)
```

```python
# 9 . A student can appear multiple times.
# Calculate the total marks for each student.
#
# Example:
# Rahul -> 50 + 70 = 120

students = [
    ("Rahul", 50),
    ("Aman", 80),
    ("Rahul", 70),
    ("Rohit", 60),
    ("Aman", 90)
]

total = {}

for name,marks in students:
    if name in total:
        total[name] += marks
    else:
        total[name] = marks
    

print(total)
```

```python
# 10 . Find all numbers that have the same frequency.

numbers = [2, 2, 3, 3, 4, 5, 5, 6, 7, 7, 8, 8, 9, 10, 10, 11]

freq = {}

# Count frequency
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

# Count how many numbers have each frequency
frequency_count = {}

for i in freq:
    f = freq[i]

    if f in frequency_count:
        frequency_count[f] += 1
    else:
        frequency_count[f] = 1

# Find numbers having a shared frequency
answer = []

for i in freq:
    if frequency_count[freq[i]] > 1:
        answer.append(i)

print(answer)
```
