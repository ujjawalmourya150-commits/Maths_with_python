```python
# 1. Create a dictionary where each number is a key
# and its square is the value.

numbers = [2, 4, 6, 8]

d = {}

for i in numbers:
    d[i]=i**2
print(d)
```

```python
# 2. Count how many times each number appears.

numbers = [2, 2, 3, 4, 3, 2]

d = {}

for i in numbers:
    
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
# 3. Create a dictionary containing only numbers
# that appear more than once.

numbers = [1, 2, 2, 3, 4, 4, 5]

d = {}
more_than_one = {}

for i in numbers:

    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i] > 1:
        more_than_one[i] = d[i]
print(more_than_one)
```

```python
# 4. Find the number with the highest frequency.
# If there is a tie, return the first one.

numbers = [1, 2, 2, 3, 4, 4, 5]

d ={}
highest = 0
highest_num = None

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i] > highest:
        highest = d[i]
        highest_num = i
print(highest_num)
```

```python
# 5 . Find the number with the highest frequency.
# If there is a tie, return the last one.

numbers = [1, 2, 2, 3, 4, 4, 5]

d ={}
highest = 0
highest_num = None

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i] >= highest:
        highest = d[i]
        highest_num = i
print(highest_num)
```

```python
# 6. Create a dictionary where each number is a key
# and its double is the value.

numbers = [3, 5, 7, 9]

d = {i:i*2 for i in numbers}  # dictionary comprehension
print(d)
```

```python
# 7. Create a dictionary where each word is a key
# and its first character is the value.

words = ["Python", "Machine", "Learning", "Artificial"]

d = {i:i[0] for i in words} # dictionary comprehension
print(d)
```

```python
# 8 .  Create a dictionary where each number is a key
# and "Even" or "Odd" is the value.

numbers = [2, 7, 10, 13, 18]

d = {i:"even" if i%2==0 else "odd" for i in numbers }

print(d)
```

```python
# 10 . Given a dictionary of students and marks,
# find the student who scored more than 80 marks using item().

students = {
    "Rahul": 72,
    "Aman": 89,
    "Rohit": 81,
    "Ujjawal": 65,
    "Karan": 92
}

for name, marks in students.items():
    if marks > 80:
        print(name)
```
