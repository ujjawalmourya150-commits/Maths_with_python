```python
# 1. Find all numbers having the SECOND-HIGHEST frequency.
#
# If multiple numbers have the same second-highest frequency,
# return all of them.

numbers = [2, 2, 2, 3, 3, 4, 4, 4, 5, 6, 6]

highest=0
second=0
highest_number=None
second_number=None

d ={}

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)

for i in d:
    if d[i] > highest:
        second = highest
        second_number = highest_number

        highest = d[i]
        highest_number = i

    elif d[i] > second and d[i] != highest:
        second = d[i]
        second_number = i

answer = []

for i in d:
    if d[i] == second:
        answer.append(i)

print("All numbers having the SECOND-HIGHEST frequency:", answer)
```

```python
# 2 . Given a list of numbers,
# find ALL numbers having the LOWEST frequency.
numbers = [7, 7, 7, 2, 2, 5, 5, 9, 9, 4, 4, 4]

d = {}

for i in numbers:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

lowest = None

for i in d:
    if lowest is None or d[i] < lowest:
        lowest = d[i]

answer = []

for i in d:
    if d[i] == lowest:
        answer.append(i)

print(f"{answer} is the lowest frequency and has frequency {lowest}")
        
```

```python
# 5 . Given a dictionary of students and marks,
# create a new dictionary where:
#
# key   -> marks
# value -> list of students who got those marks

students = {
    "Rahul": 80,
    "Aman": 90,
    "Rohit": 80,
    "Karan": 70,
    "Mohit": 90,
    "Raj": 80
}

d = {}

for name,marks in students.items():
    if marks in d:
        d[marks].append(name)
    else:
        d[marks]=[name]
print(d)
```

```python
# 8 . Given a list of words,
# find ALL words having the HIGHEST frequency.
#
# If multiple words have the same highest frequency,
# return all of them.
words = [
    "apple", "banana", "apple",
    "orange", "banana", "orange",
    "grape"
]

d = {}

for i in words:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

highest = 0
answer = []

for i in d:
    if d[i] > highest:
        highest = d[i]
        answer = [i]

    elif d[i] == highest:
        answer.append(i)

print(answer)
```

```python
# 9. Given a list of numbers,
# find the THIRD-HIGHEST DISTINCT FREQUENCY
# and return ALL numbers having that frequency.
#
# Frequencies:
# 2 -> 4 times
# 3 -> 3 times
# 4 -> 2 times
# 5 -> 2 times
# 6 -> 1 time
#
# Third-highest distinct frequency = 2
#
# Expected answer:
# [4, 5]
#
# Do NOT use:
# max(), min(), sorted(), set(), Counter
#
numbers = [
    2, 2, 2, 2,
    3, 3, 3,
    4, 4,
    5, 5,
    6
]

d ={}
largest = 0
second = 0
third = 0

for i in numbers:
    if i in d:
        d[i]+=1
    else:
        d[i]=1

for i in d:
    if d[i]>largest:
        third=second
        second=largest
        largest=d[i]
      
    elif d[i]>second and d[i]!=largest:
        third=second
        second=d[i]
        
    elif d[i]>third and d[i]!=second and d[i]!=largest:
        third=d[i]
d2=[]
for i in d:
    if d[i]==third:
        d2.append(i)

print(d2)
```

```python
# 10 . Given a list of numbers,
# find the FIRST number that appears exactly
# as many times as the length of the list of unique numbers.
#
# Example:
# numbers = [2, 2, 3, 3, 3, 4, 5]
#
# Unique numbers = 4
# So find the first number that appears 4 times.

numbers = [2, 2, 2, 2, 3, 3, 4, 5]

d = {}

for i in numbers:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

unique_count = len(d)

answer = None

for i in numbers:
    if d[i] == unique_count:
        answer = i
        break

print(answer)
```
