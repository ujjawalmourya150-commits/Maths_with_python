```python
# 1 .WAP to create list of prime number from 1 to 100

Prime_number = []

for num in range(2,101):
    count = 0
    for i in range(1,num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        Prime_number.append(num)
print(Prime_number)
```

```python
# 5  .Separte positive and nagative

numbers = [1,4,3,5,6,-1,-2,-3,-6,0]

Positive = []
Negative = []
Zero = []

for i in numbers:
    if i > 0:
        Positive.append(i)
    elif i < 0:
        Negative.append(i)
    else :
        Zero.append(i)
print(f"Positive list is {Positive}")
print(f"Negative list is {Negative}")
print(f"Zero list is {Zero}")

        
```

```python
# 6 . Move Zero to end 

new_list = []

for i in numbers:
    if i != 0:
        new_list.append(i)

for i in numbers:
    if i == 0:
        new_list.append(i)

print(new_list)
```

```python
# 7 . find Missing number 

numbers = [1,2,3,5,6,7,8]

for i in range(1,len(numbers)+1):
    if i not in numbers:
        print(i)
```

```python
# 8 . Find Duplicate Numbers

numbers = [4, 7, 2, 4, 9, 7, 5, 2, 8]

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] == numbers[j]:
            print(numbers[i])
            break
```

```python
# 9. Find local maximum

numbers = [4, 7, 2, 4, 9, 7, 5, 2, 8]

for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        print("Local maximum:", numbers[i])
```

```python
# 10 . Create a new list for odd and even

num = [3,2,4,5,9,7,1,6,11]
print(f"original list------------->{num}")
Even_list = []
Odd_list = []
for i in num:
    if i % 2 == 0:
        Even_list.append(i)
    else:
        Odd_list.append(i)
print(f"Even list ---------> {Even_list}")
print(f"Odd list-----------> {Odd_list}")
```
