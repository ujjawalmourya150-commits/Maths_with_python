```python
#  1 .Shopping discount 

shopping_amount = int(input("Enter shopping amount : "))

discount_20 = shopping_amount * (20/100)
discount_15 = shopping_amount * (15/100)
discount_10 = shopping_amount * (10/100)

if shopping_amount >= 5000:
    print(f"You get 20 % discount and your total bill is {shopping_amount - discount_20 }")
elif shopping_amount >= 3000 and shopping_amount < 5000 :
    print(f"You get 15 % discount and your total bill is {shopping_amount - discount_15}")
elif shopping_amount >= 1000 and shopping_amount < 3000:
    print(f"You get 10 % discount and your total bill is {shopping_amount - discount_10}")
else:
    print(f"No discount and your total bill is {shopping_amount}")
```

```python
# 4 .  Date Validity 

day = int(input("Enter the day : "))
month = int(input("Enter the month : "))
year = int(input("Enter the year : "))

days_30 = [4,6,9,11]

if month < 1 or month > 12:
    print("Invalid date ")
elif month == 2:
    if year % 400 == 0 or year % 4 == 0 and year % 100 != 0:
        if day >= 1 and day <=29:
            print("Valid date")
        else:
            print("Invalid date")
    else :
        if day >= 1 and day <=28:
            print("Valid date")
        else:
            print("Invalid date")
elif month in days_30:
    if day >= 1 and day <=30:
        print("Valid date")
    else:
        print("Invalid date")
else:
    if day >= 1 and day <=31:
        print("Valid date")
    else:
        print("Invalid date")
```

```python
# 5 . Triangle Validity

a = int(input("Enter the first side of the triangle : "))
b = int(input("Enter the second side of the triangle : "))
c = int(input("Enter the third side of the triangle : "))

if a>b and a>c:
    maxi = a
elif b>a and b>c:
    maxi = b
else:
    maxi = c

if maxi > (a+b+c-maxi):
    print("Triangle can't be formed")
else:
    if a == b == c:
        print("Triangle is an eqivalent triangle") 
    elif a==b or b==c or a==c:
        print("Triangle is an Isosceles triangle")
    else:
        print("Triangle is an Scalene triangle")
```

```python
# 6 . Electricity Bill

unit = float(input("Enter your electricity units : "))

if unit <= 100:
    bill = unit * 5

elif unit <= 200:
    bill = (100 * 5) + ((unit - 100) * 7)

else:
    bill = (100 * 5) + (100 * 7) + ((unit - 200) * 10)

if bill > 2000:
    surcharge = bill * 5 / 100
    bill = bill + surcharge

print(f"Your electricity bill is ₹{bill}")
```

```python
# 9 . Middle Numbers with 4 inputs

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
d = int(input("Enter fourth number: "))

if a < b and a < c and a < d:
    smallest = a
elif b < a and b < c and b < d:
    smallest = b
elif c < a and c < b and c < d:
    smallest = c
else:
    smallest = d

if a > b and a > c and a > d:
    largest = a
elif b > a and b > c and b > d:
    largest = b
elif c > a and c > b and c > d:
    largest = c
else:
    largest = d

if a != smallest and a != largest:
    print("Middle number:", a)

if b != smallest and b != largest:
    print("Middle number:", b)

if c != smallest and c != largest:
    print("Middle number:", c)

if d != smallest and d != largest:
    print("Middle number:", d)
```
